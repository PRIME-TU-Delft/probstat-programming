# Critical Region

In this section we will be looking at the critical region in a hypothesis test. Let us consider the distribution $T \sim Bin(1000,0.01)$ with $\alpha=0.05$. We start with a one-tailed test. Our hypotheses are as follows:

$$
\begin{aligned}
H_0 &: \theta = 0.01 \\
H_1 &: \theta > 0.01 \\
\end{aligned}
$$

In the code below we simulate this binomial distribution. We use scipy's `binom.pmf`to get the PMF of our distribution and calculate our critical values using `binom.ppf`.

The critical region is the area where the p-value, $P_{H_0}(T \geq x)$ is less than the $\alpha$ value. This region is highlighted in our plot. Run the code below to see the critical region and how it is impacted by different $\alpha$ values. This critical region indicates the range of values of $T$ where we would reject the null hypothesis.

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interactive, fixed
from scipy.stats import norm, ecdf, expon, binom
import seaborn as sns
```

```python
def update_binomial(n, p, sample_size, alpha_val):
    r_vals = np.arange(n + 1)
    rng = np.random.default_rng()
    dist_binomial = binom.pmf(r_vals, n, p)
    sample = rng.binomial(n, p, size=sample_size)
    freqs = np.bincount(sample.astype(np.int32), minlength=n + 1) / sample_size
    critical_high = binom.ppf(1 - alpha_val, n, p)
    
    colors = ['blue'] * len(r_vals)
    
    for i, r in enumerate(r_vals):
        if r >= critical_high:
            colors[i] = 'red'  
    
    plt.figure(figsize=(8, 5))
    # plt.bar(r_vals, freqs, alpha=0.7, label="Sampled Frequency", color=colors)
    plt.bar(r_vals, dist_binomial, alpha=0.7, label="Sampled Frequency", color=colors)
    plt.xlim(0,24)
    plt.xlabel("r")
    plt.ylabel("Probability")
    plt.title("Critical Region for one-tailed Binomial Test")
    plt.legend()
    plt.show()

n = 1000
p = 0.01
size = 200

slider_alpha = widgets.FloatSlider(value=0.05, min=0.01, max=0.2, step=0.01, description="alpha value")
interactive_plot = interactive(
    update_binomial,
    n=fixed(n),
    p=fixed(p),
    sample_size=fixed(size),
    alpha_val=slider_alpha
)
interactive_plot
```

Now let us look at the same distribution but using a two tail test. In this case our hypotheses are:

$$
\begin{aligned}
H_0 &: \theta = 0.01 \\
H_1 &: \theta \ne 0.01 \\
\end{aligned}
$$


```python
def update_binomial(n, p, sample_size, alpha_val):
    r_vals = np.arange(n + 1)
    rng = np.random.default_rng()
    dist_binomial = binom.pmf(r_vals, n, p)
    sample = rng.binomial(n, p, size=sample_size)
    freqs = np.bincount(sample.astype(np.int32), minlength=n + 1) / sample_size
    critical_high = binom.ppf(1 - alpha_val/2, n, p)
    critical_low = binom.ppf(alpha_val/2, n, p)
    
    colors = ['blue'] * len(r_vals)
    
    for i, r in enumerate(r_vals):
        if r >= critical_high or r <= critical_low:
            colors[i] = 'red'  
    
    plt.figure(figsize=(8, 5))
    # plt.bar(r_vals, freqs, alpha=0.7, label="Sampled Frequency", color=colors)
    plt.bar(r_vals, dist_binomial, alpha=0.7, label="Sampled Frequency", color=colors)
    plt.xlim(0,24)
    plt.xlabel("r")
    plt.ylabel("Probability")
    plt.title("Critical Region for two-tailed Binomial Test")
    plt.legend()
    plt.show()

n = 1000
p = 0.01
size = 200

slider_alpha = widgets.FloatSlider(value=0.05, min=0.01, max=0.2, step=0.01, description="alpha value")
interactive_plot = interactive(
    update_binomial,
    n=fixed(n),
    p=fixed(p),
    sample_size=fixed(size),
    alpha_val=slider_alpha
)
interactive_plot
```
