# Power and Significance

In this section we will be exploring power curves and one-sided vs two-sided tests. 

## Power Curves 

Power curves are a graph of power versus all values of $\mu$. We will start with an example using a z-test. In order to use this z-test, we will use the population variance. Run the code below to get an example of a right-sided test. 

````python
import micropip

await micropip.install("ipywidgets")
````

````python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interactive, fixed
from scipy import stats
````

````python
def right_sided_ztest_power_curve(mu1_minus_mu0, sigma, n, alpha):
    plt.figure(figsize=(10, 6))

    # Plotting the power curve for different alpha levels
    for a in alpha:
        # Calculate the critical z-value for the given alpha
        z_crit = stats.norm.ppf(1 - a)
        powers = []
        # Calculate the power for each effect size
        for d in mu1_minus_mu0:
            z_beta = (d * np.sqrt(n) / sigma) - z_crit
            power = stats.norm.cdf(z_beta)
            powers.append(power)
        # Plot the power curve
        plt.plot(mu1_minus_mu0, powers, label=f"Alpha = {a}")
    plt.xlabel("Effect Size ($\mu_1 - \mu_0$)")
    plt.ylabel("Power")
    plt.title("Power Curve for Right-Sided Z-Test")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()


alpha = np.array([0.1, 0.05, 0.025, 0.01, 0.005])

n_vals = widgets.IntSlider(
    value=100, min=20, max=200, step=10, description="Max Sample Size"
)
sigma_vals = widgets.FloatSlider(
    value=2.55, min=0.1, max=5.0, step=0.1, description="Standard Deviation"
)
mu1_minus_mu0_vals = np.linspace(0, 2, 100)
interactive_plot = interactive(
    right_sided_ztest_power_curve,
    mu1_minus_mu0=fixed(mu1_minus_mu0_vals),
    sigma=sigma_vals,
    n=n_vals,
    alpha=fixed(alpha),
)
interactive_plot
````

Now run the code block below to plot the power curves for a left-sided test. Observe how the z-value used in the right sided test corresponds to a probability of $1-\alpha$ in the left tail of the standard normal distribution (which is $\alpha$ for the right tail), whereas the z-value used in the left-sided test uses a value corresponding to $\alpha$ in the left tail of the standard normal distribution.

````python
def left_sided_ztest_power_curve(mu1_minus_mu0, sigma, n, alpha):
    plt.figure(figsize=(10, 6))
    # Plotting the power curve for different alpha levels
    for a in alpha:
        # Calculate the critical z-value for the given alpha
        z_crit = stats.norm.ppf(a)
        powers = []
        # Calculate the power for each effect size
        for d in mu1_minus_mu0:
            z_beta = z_crit - (d * np.sqrt(n) / sigma)
            power = stats.norm.cdf(z_beta)
            powers.append(power)
        # Plot the power curve
        plt.plot(mu1_minus_mu0, powers, label=f"Alpha = {a}")
    plt.xlabel("Effect Size")
    plt.ylabel("Power")
    plt.title("Power Curve for Left-Sided Z-Test")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()


alpha = np.array([0.1, 0.05, 0.025, 0.01, 0.005])

n_vals = widgets.IntSlider(
    value=100, min=20, max=200, step=10, description="Max Sample Size"
)
sigma_vals = widgets.FloatSlider(
    value=2.55, min=0.1, max=5.0, step=0.1, description="Standard Deviation"
)
mu1_minus_mu0_vals = np.linspace(-2, 0.5, 100)
interactive_plot = interactive(
    left_sided_ztest_power_curve,
    mu1_minus_mu0=fixed(mu1_minus_mu0_vals),
    sigma=sigma_vals,
    n=n_vals,
    alpha=fixed(alpha),
)
interactive_plot
````

Finally, let's look at an example of a two-sided test. Run the code below.

````python
def two_sided_ztest_power_curve(mu1_minus_mu0, sigma, n, alpha):
    plt.figure(figsize=(10, 6))
    # Plotting the power curve for different alpha levels
    for a in alpha:
        # Calculate the critical z-value for the given alpha
        z_crit = stats.norm.ppf(1 - a / 2)
        powers = []
        # Calculate the power for each effect size
        for d in mu1_minus_mu0:
            upper_power = (d * np.sqrt(n) / sigma) - z_crit
            lower_power = -1.0 * (d * np.sqrt(n) / sigma) - z_crit
            power = stats.norm.cdf(upper_power) + stats.norm.cdf(lower_power)
            powers.append(power)
        # Plot the power curve
        plt.plot(mu1_minus_mu0, powers, label=f"Alpha = {a}")
    plt.xlabel("Effect Size")
    plt.ylabel("Power")
    plt.title("Power Curve for Two-Sided Z-Test")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()


alpha = np.array([0.1, 0.05, 0.025, 0.01, 0.005])

n_vals = widgets.IntSlider(
    value=100, min=20, max=200, step=10, description="Max Sample Size"
)
sigma_vals = widgets.FloatSlider(
    value=2.55, min=0.1, max=5.0, step=0.1, description="Standard Deviation"
)
mu1_minus_mu0_vals = np.linspace(-2, 2, 200)
interactive_plot = interactive(
    two_sided_ztest_power_curve,
    mu1_minus_mu0=fixed(mu1_minus_mu0_vals),
    sigma=sigma_vals,
    n=n_vals,
    alpha=fixed(alpha),
)
interactive_plot
````

```{note}
The z-test setting we are using in this example is relevant when we have a known population variance. In the setting where $\sigma^2$ is not known we will want to be using a t-test. When using the t-test we can estimate $\sigma^2$ using the sample standard deviation of a pre-sample or the sample itself.

Now let us move onto a different test. In this scenario we will be using a Poisson distribution and use the sum as the test statistic. The power will be dependent on the effect size ($\mu_1 - \mu_0$), $n$ and $\alpha$. Run the code below to simulate a right, left, and two-sided test.

````python
def right_sided_poisson_test_power_curve(mu1_minus_mu0, mu1, n, alpha):
    plt.figure(figsize=(10, 6))
    # Plotting the power curve for different alpha levels
    for a in alpha:
        powers = []
        # Calculate the power for each effect size
        for d in mu1_minus_mu0:
            mu0 = mu1 - d

            if mu0 <= 0:
                powers.append(np.nan)
                continue

            k_crit = stats.poisson.ppf(1 - a, n * mu0)

            power = 1 - stats.poisson.cdf(k_crit - 1, n * mu1)
            powers.append(power)
        # Plot the power curve
        plt.plot(mu1_minus_mu0, powers, label=f"Alpha = {a}", linewidth=2)

    plt.xlabel("Effect Size")
    plt.ylabel("Power")
    plt.title("Power Curve for Right-Sided Test")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()


alpha = np.array([0.1, 0.05, 0.025, 0.01, 0.005])

mu1_slider = widgets.FloatSlider(
    value=3.0, min=0.5, max=10.0, step=0.1, description="µ1"
)

n_slider = widgets.IntSlider(value=50, min=10, max=200, step=5, description="n")

mu1_minus_mu0_vals = np.linspace(-4, 4, 200)

interactive_plot = interactive(
    right_sided_poisson_test_power_curve,
    mu1_minus_mu0=fixed(mu1_minus_mu0_vals),
    mu1=mu1_slider,
    n=n_slider,
    alpha=fixed(alpha),
)
interactive_plot
````

````python
def left_sided_poisson_test_power_curve(mu1_minus_mu0, mu1, n, alpha):
    plt.figure(figsize=(10, 6))
    # Plotting the power curve for different alpha levels
    for a in alpha:
        powers = []
        # Calculate the power for each effect size
        for d in mu1_minus_mu0:
            mu0 = mu1 - d

            if mu0 <= 0:
                powers.append(np.nan)
                continue

            k_crit = stats.poisson.ppf(a, n * mu0)

            power = stats.poisson.cdf(k_crit - 1, n * mu1)
            powers.append(power)
        # Plot the power curve
        plt.plot(mu1_minus_mu0, powers, label=f"Alpha = {a}", linewidth=2)

    plt.xlabel("Effect Size")
    plt.ylabel("Power")
    plt.title("Power Curve for Left-Sided Test")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()


alpha = np.array([0.1, 0.05, 0.025, 0.01, 0.005])

mu1_slider = widgets.FloatSlider(
    value=3.0, min=0.5, max=10.0, step=0.1, description="µ1"
)

n_slider = widgets.IntSlider(value=50, min=10, max=200, step=5, description="n")

mu1_minus_mu0_vals = np.linspace(-4, 4, 200)

interactive_plot = interactive(
    left_sided_poisson_test_power_curve,
    mu1_minus_mu0=fixed(mu1_minus_mu0_vals),
    mu1=mu1_slider,
    n=n_slider,
    alpha=fixed(alpha),
)
interactive_plot
````

````python
def two_sided_poisson_test_power_curve(mu1_minus_mu0, mu1, n, alpha):
    plt.figure(figsize=(10, 6))
    # Plotting the power curve for different alpha levels
    for a in alpha:
        powers = []
        # Calculate the power for each effect size
        for d in mu1_minus_mu0:
            mu0 = mu1 - d

            if mu0 <= 0:
                powers.append(np.nan)
                continue

            k_lower = stats.poisson.ppf(a / 2, n * mu0)
            k_upper = stats.poisson.ppf(1 - a / 2, n * mu0)

            power_lower = stats.poisson.cdf(k_lower, n * mu1)
            power_upper = 1 - stats.poisson.cdf(k_upper - 1, n * mu1)
            power = power_lower + power_upper
            powers.append(power)
        # Plot the power curve
        plt.plot(mu1_minus_mu0, powers, label=f"Alpha = {a}", linewidth=2)

    plt.xlabel("Effect Size")
    plt.ylabel("Power")
    plt.title("Power Curve for Two-Sided Z-Test")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()


alpha = np.array([0.1, 0.05, 0.025, 0.01, 0.005])

mu1_slider = widgets.FloatSlider(
    value=3.0, min=0.5, max=10.0, step=0.1, description="µ1"
)

n_slider = widgets.IntSlider(value=50, min=10, max=200, step=5, description="n")

mu1_minus_mu0_vals = np.linspace(-4, 4, 200)

interactive_plot = interactive(
    two_sided_poisson_test_power_curve,
    mu1_minus_mu0=fixed(mu1_minus_mu0_vals),
    mu1=mu1_slider,
    n=n_slider,
    alpha=fixed(alpha),
)
interactive_plot
````

### Bonus: Visualisation

In this section we will visualize the type I and type II errors associated with the two examples given above. We will be doing so by plotting the PDF of test statistics under $H_0$ and the PDF of test statistics under $H_1$. The shaded regions correspond to the type I and type II errors.

````python
def right_sided_ztest_pdfs(mu1_minus_mu0, sigma, n, alpha):
    fig, ax = plt.subplots(figsize=(12, 8))

    z_range = np.linspace(-8, 8, 1000)
    # Calculate the PDFs for the null and alternative hypotheses
    h0_pdf = stats.norm.pdf(z_range, loc=0, scale=1)

    standardized_effect = mu1_minus_mu0 * np.sqrt(n) / sigma
    h1_pdf = stats.norm.pdf(z_range, loc=standardized_effect, scale=1)

    # Plot the PDFs for H0 and H1
    ax.plot(z_range, h0_pdf, label="H_0", color="blue", linewidth=2)
    ax.plot(
        z_range,
        h1_pdf,
        label=f"H_1: mean = {mu1_minus_mu0:.2f}",
        color="orange",
        linewidth=2,
    )

    z_crit = stats.norm.ppf(1 - alpha)
    # Add critical value line and shaded regions for Type I and Type II errors
    ax.axvline(
        z_crit,
        color="red",
        linestyle="--",
        linewidth=2,
        label=f"Critical value = {z_crit:.3f}",
    )

    # Shaded region for Type I error (H0 rejected when it is true)
    z_alpha_region = z_range[z_range >= z_crit]
    h0_alpha_region = stats.norm.pdf(z_alpha_region, loc=0, scale=1)
    ax.fill_between(
        z_alpha_region,
        0,
        h0_alpha_region,
        color="blue",
        alpha=0.3,
        label=f"Type I Error = {alpha:.3f}",
    )

    # Shaded region for Type II error (H1 not rejected when it is true)
    z_beta_region = z_range[z_range < z_crit]
    h1_beta_region = stats.norm.pdf(z_beta_region, loc=standardized_effect, scale=1)
    beta = stats.norm.cdf(z_crit, loc=standardized_effect, scale=1)
    ax.fill_between(
        z_beta_region,
        0,
        h1_beta_region,
        color="orange",
        alpha=0.3,
        label=f"Type II Error = {beta:.3f}",
    )

    ax.set_title("Right-Sided Z-Test PDFs for Normal Distribution", fontsize=16)
    ax.set_xlabel("Test Statistic (Z-score)", fontsize=12)
    ax.set_ylabel("Probability Density", fontsize=12)
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_ylim(0, max(max(h0_pdf), max(h1_pdf)) * 1.1)

    plt.tight_layout()
    plt.show()


alpha = widgets.FloatSlider(
    value=0.01, min=0.005, max=0.1, step=0.005, description="Alpha"
)

n_vals = widgets.IntSlider(value=100, min=20, max=200, step=10, description="n")
sigma_vals = widgets.FloatSlider(
    value=2.55, min=0.1, max=5.0, step=0.1, description="Std Deviation"
)
mu1_minus_mu0_vals = widgets.FloatSlider(
    value=1, min=0, max=2, step=0.1, description="Effect Size"
)
interactive_plot = interactive(
    right_sided_ztest_pdfs,
    mu1_minus_mu0=mu1_minus_mu0_vals,
    sigma=sigma_vals,
    n=n_vals,
    alpha=alpha,
)
interactive_plot
````

````python
def right_sided_ztest_pdfs_poisson(mu1_minus_mu0, mu1, n, alpha):
    fig, ax = plt.subplots(figsize=(12, 8))

    mu0 = mu1 - mu1_minus_mu0
    # Check if mu0 is valid for Poisson distribution
    if mu0 <= 0:
        ax.text(
            0.5,
            0.5,
            f"Invalid: mu = {mu0:.2f} <= 0\nPoisson parameter must be positive",
            transform=ax.transAxes,
            ha="center",
            va="center",
            fontsize=16,
            color="red",
        )
        ax.set_title("Invalid Parameter Range")
        plt.show()
        return

    lambda0 = n * mu0
    lambda1 = n * mu1

    std0 = np.sqrt(lambda0)
    std1 = np.sqrt(lambda1)

    x_min0 = max(0, int(lambda0 - 5 * std0))
    x_max0 = int(lambda0 + 5 * std0)
    x_min1 = max(0, int(lambda1 - 5 * std1))
    x_max1 = int(lambda1 + 5 * std1)
    x_min = min(x_min0, x_min1)
    x_max = max(x_max0, x_max1)

    # Ensure the range is wide enough to visualize the distributions
    if x_max - x_min < 20:
        center = (x_min + x_max) / 2
        x_min = max(0, int(center - 10))
        x_max = int(center + 10)

    x_range = np.arange(x_min, x_max + 1)

    # Calculate the PMFs for the null and alternative hypotheses
    h0_pmf = stats.poisson.pmf(x_range, lambda0)
    h1_pmf = stats.poisson.pmf(x_range, lambda1)

    # Plot the PMFs for H0 and H1
    width = 0.4
    ax.bar(
        x_range - width / 2,
        h0_pmf,
        width=width,
        label=f"H_0: lambda = {lambda0:.1f}",
        color="blue",
        alpha=0.3,
    )
    ax.bar(
        x_range + width / 2,
        h1_pmf,
        width=width,
        label=f"H_1: lambda = {lambda1:.1f}",
        color="orange",
        alpha=0.3,
    )

    # Add critical value line and shaded regions for Type I and Type II errors
    k_crit = stats.poisson.ppf(1 - alpha, lambda0)

    ax.axvline(
        k_crit + 0.5,
        color="red",
        linestyle="--",
        linewidth=2,
        label=f"Critical value = {int(k_crit)}",
    )

    # Shaded region for Type I error (H0 rejected when it is true)
    actual_alpha = 1 - stats.poisson.cdf(k_crit - 1, lambda0)
    x_alpha = x_range[x_range >= k_crit]
    h0_alpha = stats.poisson.pmf(x_alpha, lambda0)
    ax.bar(
        x_alpha - width / 2,
        h0_alpha,
        width=width,
        color="cyan",
        alpha=1.0,
        label=f"Type I Error = {actual_alpha:.4f}",
    )

    # Shaded region for Type II error (H1 not rejected when it is true)
    beta = stats.poisson.cdf(k_crit - 1, lambda1)
    x_beta = x_range[x_range < k_crit]
    h1_beta = stats.poisson.pmf(x_beta, lambda1)
    ax.bar(
        x_beta + width / 2,
        h1_beta,
        width=width,
        color="green",
        alpha=1.0,
        label=f"Type II Error = {beta:.4f}",
    )

    ax.set_title("Right-Sided Z-Test PDFs for Poisson Distribution", fontsize=16)
    ax.set_xlabel("Test Statistic (Sum of n observations)", fontsize=12)
    ax.set_ylabel("Probability Mass", fontsize=12)

    ax.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
    ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()


alpha_slider = widgets.FloatSlider(
    value=0.05, min=0.01, max=0.1, step=0.01, description="Alpha"
)
n_slider = widgets.IntSlider(value=50, min=10, max=100, step=5, description="n")
mu1_slider = widgets.FloatSlider(
    value=3.0, min=0.5, max=8.0, step=0.1, description="true mean"
)
effect_slider = widgets.FloatSlider(
    value=1.0, min=0, max=2.0, step=0.1, description="Effect Size"
)

interactive_plot = interactive(
    right_sided_ztest_pdfs_poisson,
    mu1_minus_mu0=effect_slider,
    mu1=mu1_slider,
    n=n_slider,
    alpha=alpha_slider,
)
interactive_plot
````

## One-sided vs Two-sided Tests

In this section we will be comparing one-sided and two-sided tests in terms of significance and power.

Run the code block below to overlay the power curves for left, right and two-sided tests. Observe how for one-sided tests, the power is greater than for the two-sided test.

````python
def ztest_power_curves(mu1_minus_mu0, sigma, n, alpha):
    plt.figure(figsize=(10, 6))

    # Calculate the critical z-values for the given alpha
    z_crit = stats.norm.ppf(1 - alpha)
    z_crit_l = stats.norm.ppf(alpha)
    z_crit_double = stats.norm.ppf(1 - alpha / 2)

    powers_right = []
    powers_left = []
    powers_two_sided = []
    for d in mu1_minus_mu0:
        # Right-sided power
        z_beta = (d * np.sqrt(n) / sigma) - z_crit
        power = stats.norm.cdf(z_beta)
        powers_right.append(power)

        # Left-sided power
        z_beta = z_crit_l - (d * np.sqrt(n) / sigma)
        power = stats.norm.cdf(z_beta)
        powers_left.append(power)

        # Two-sided power
        upper_power = (d * np.sqrt(n) / sigma) - z_crit_double
        lower_power = -1.0 * (d * np.sqrt(n) / sigma) - z_crit_double
        power = stats.norm.cdf(upper_power) + stats.norm.cdf(lower_power)
        powers_two_sided.append(power)

    plt.plot(
        mu1_minus_mu0,
        powers_left,
        label="Left-Sided Test",
        linestyle="--",
        color="blue",
    )
    plt.plot(
        mu1_minus_mu0,
        powers_right,
        label="Right-Sided Test",
        linestyle="-",
        color="orange",
    )
    plt.plot(
        mu1_minus_mu0,
        powers_two_sided,
        label="Two-Sided Test",
        linestyle=":",
        color="green",
    )
    plt.xlabel("Effect Size")
    plt.ylabel("Power")
    plt.title(f"Power Curve for Z-Test with alpha={alpha:.3f}")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()


alpha = widgets.FloatSlider(
    value=0.05, min=0.005, max=0.1, step=0.005, description="Alpha"
)

n_vals = widgets.IntSlider(
    value=100, min=20, max=200, step=10, description="Max Sample Size"
)
sigma_vals = widgets.FloatSlider(
    value=2.55, min=0.1, max=5.0, step=0.1, description="Standard Deviation"
)
mu1_minus_mu0_vals = np.linspace(-2, 2, 200)
interactive_plot = interactive(
    ztest_power_curves,
    mu1_minus_mu0=fixed(mu1_minus_mu0_vals),
    sigma=sigma_vals,
    n=n_vals,
    alpha=alpha,
)
interactive_plot
````

Next let us look at the significance level. Run the code block below to visualise the critical regions for one sided and two-sided tests.

````python
def ztest_pdfs(mu1_minus_mu0, sigma, n, alpha):
    fig, ax = plt.subplots(figsize=(12, 8))

    z_range = np.linspace(-8, 8, 1000)

    # Calculate the PDFs for the null and alternative hypotheses
    h0_pdf = stats.norm.pdf(z_range, loc=0, scale=1)

    standardized_effect = mu1_minus_mu0 * np.sqrt(n) / sigma
    h1_pdf = stats.norm.pdf(z_range, loc=standardized_effect, scale=1)

    # Plot the PDFs for H0 and H1
    ax.plot(z_range, h0_pdf, label="H_0", color="blue", linewidth=2)
    ax.plot(
        z_range,
        h1_pdf,
        label=f"H_1: mean = {mu1_minus_mu0:.2f}",
        color="orange",
        linewidth=2,
    )

    # Critical values for right-sided test
    z_crit_r = stats.norm.ppf(1 - alpha)

    ax.axvline(
        z_crit_r,
        color="red",
        linestyle="--",
        linewidth=2,
        label=f"Critical value right sided = {z_crit_r:.3f}",
    )

    # Shaded region for Type I error (H0 rejected when it is true)
    z_alpha_region = z_range[z_range >= z_crit_r]
    h0_alpha_region = stats.norm.pdf(z_alpha_region, loc=0, scale=1)
    ax.fill_between(
        z_alpha_region,
        0,
        h0_alpha_region,
        color="red",
        alpha=0.3,
        label=f"Type I Error = {alpha:.3f}",
    )

    # Critical values for two-sided test
    z_crit_two_sided = stats.norm.ppf(1 - alpha / 2)

    ax.axvline(
        z_crit_two_sided,
        color="blue",
        linestyle="--",
        linewidth=2,
        label=f"Critical value two-sided = {z_crit_two_sided:.3f}",
    )

    ax.axvline(
        -z_crit_two_sided,
        color="blue",
        linestyle="--",
        linewidth=2,
        label=f"Critical value two-sided = -{z_crit_two_sided:.3f}",
    )

    # Shaded region for Type I error (H0 rejected when it is true) for two-sided test
    z_alpha_region = z_range[z_range >= z_crit_two_sided]
    z_alpha_region = np.append(z_alpha_region, z_range[z_range <= -z_crit_two_sided])
    h0_alpha_region = stats.norm.pdf(z_alpha_region, loc=0, scale=1)
    ax.fill_between(
        z_alpha_region,
        0,
        h0_alpha_region,
        color="blue",
        alpha=0.3,
        label=f"Type I Error = {alpha:.3f}",
    )

    ax.set_title("Z-Test PDFs for Normal Distribution", fontsize=16)
    ax.set_xlabel("Test Statistic (Z-score)", fontsize=12)
    ax.set_ylabel("Probability Density", fontsize=12)
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_ylim(0, max(max(h0_pdf), max(h1_pdf)) * 1.1)

    plt.tight_layout()
    plt.show()


alpha = widgets.FloatSlider(
    value=0.01, min=0.005, max=0.1, step=0.005, description="Alpha"
)

n_vals = widgets.IntSlider(value=100, min=20, max=200, step=10, description="n")
sigma_vals = widgets.FloatSlider(
    value=2.55, min=0.1, max=5.0, step=0.1, description="Std Deviation"
)
mu1_minus_mu0_vals = widgets.FloatSlider(
    value=1, min=-2, max=2, step=0.1, description="Effect Size"
)
interactive_plot = interactive(
    ztest_pdfs,
    mu1_minus_mu0=mu1_minus_mu0_vals,
    sigma=sigma_vals,
    n=n_vals,
    alpha=alpha,
)
interactive_plot
````

Finally, we can analyse the difference in how one-sided and two-sided tests behave depending on p-value.

````python
def ztest_pdfs(mu1_minus_mu0, sigma, n, alpha, crit_val):
    fig, ax = plt.subplots(figsize=(12, 8))

    z_range = np.linspace(-8, 8, 1000)

    # Calculate the PDFs for the null and alternative hypotheses
    h0_pdf = stats.norm.pdf(z_range, loc=0, scale=1)

    standardized_effect = mu1_minus_mu0 * np.sqrt(n) / sigma
    h1_pdf = stats.norm.pdf(z_range, loc=standardized_effect, scale=1)

    # Plot the PDFs for H0 and H1
    ax.plot(z_range, h0_pdf, label="H_0", color="blue", linewidth=2)
    ax.plot(
        z_range,
        h1_pdf,
        label=f"H_1: mean = {mu1_minus_mu0:.2f}",
        color="orange",
        linewidth=2,
    )

    # Add critical value line
    ax.axvline(
        crit_val,
        color="cyan",
        linestyle="--",
        linewidth=2,
        label=f"Observed test statistic = {crit_val:.3f}",
    )

    # Shaded region for p-value
    z_alpha_region = z_range[z_range >= crit_val]
    h0_alpha_region = stats.norm.pdf(z_alpha_region, loc=0, scale=1)
    ax.fill_between(
        z_alpha_region,
        0,
        h0_alpha_region,
        color="cyan",
        alpha=0.3,
    )

    # Critical values for right-sided test
    p_val_r = stats.norm.sf(crit_val)
    z_crit_r = stats.norm.ppf(1 - alpha)

    ax.axvline(
        z_crit_r,
        color="red",
        linestyle="--",
        linewidth=2,
        label=f"Critical value right sided = {z_crit_r:.3f}",
    )

    z_alpha_region = z_range[z_range >= z_crit_r]
    h0_alpha_region = stats.norm.pdf(z_alpha_region, loc=0, scale=1)
    ax.fill_between(
        z_alpha_region,
        0,
        h0_alpha_region,
        color="red",
        alpha=0.3,
        label=f"Critical Region right sided",
    )

    # Critical values for two-sided test
    p_val_double_sided = stats.norm.sf(crit_val) * 2
    z_crit_two_sided = stats.norm.ppf(1 - alpha / 2)

    ax.axvline(
        z_crit_two_sided,
        color="blue",
        linestyle="--",
        linewidth=2,
    )

    ax.axvline(
        -z_crit_two_sided,
        color="blue",
        linestyle="--",
        linewidth=2,
        label=f"Critical value two-sided = {z_crit_two_sided:.3f}",
    )

    z_alpha_region = z_range[z_range >= z_crit_two_sided]
    z_alpha_region = np.append(z_alpha_region, z_range[z_range <= -z_crit_two_sided])
    h0_alpha_region = stats.norm.pdf(z_alpha_region, loc=0, scale=1)
    ax.fill_between(
        z_alpha_region,
        0,
        h0_alpha_region,
        color="blue",
        alpha=0.3,
        label=f"Critical Region two-sided",
    )

    print(f"p-value for right-sided test: {p_val_r:.4f}")
    if p_val_r < alpha:
        print("Reject H0 in right-sided test")
    else:
        print("Do not reject H0 in right-sided test")
    print(f"p-value for two-sided test: {p_val_double_sided:.4f}")
    if p_val_double_sided < alpha:
        print("Reject H0 in two-sided test")
    else:
        print("Do not reject H0 in two-sided test")

    ax.set_title("Z-Test PDFs for Normal Distribution", fontsize=16)
    ax.set_xlabel("Test Statistic (Z-score)", fontsize=12)
    ax.set_ylabel("Probability Density", fontsize=12)
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_ylim(0, max(max(h0_pdf), max(h1_pdf)) * 1.1)

    plt.tight_layout()
    plt.show()


alpha = widgets.FloatSlider(
    value=0.05, min=0.005, max=0.1, step=0.005, description="Alpha"
)

n_vals = widgets.IntSlider(value=100, min=20, max=200, step=10, description="n")
sigma_vals = widgets.FloatSlider(
    value=2.55, min=0.1, max=5.0, step=0.1, description="Std Deviation"
)
mu1_minus_mu0_vals = widgets.FloatSlider(
    value=1, min=-2, max=2, step=0.1, description="Effect Size"
)

test_statistic_vals = widgets.FloatSlider(
    value=0, min=-5, max=5, step=0.1, description="Test Statistic"
)

interactive_plot = interactive(
    ztest_pdfs,
    mu1_minus_mu0=mu1_minus_mu0_vals,
    sigma=sigma_vals,
    n=n_vals,
    alpha=alpha,
    crit_val=test_statistic_vals,
)
interactive_plot
````

As can be observed from the above examples, we reject the null hypothesis more quickly for a one-sided case than a two-sided case.
