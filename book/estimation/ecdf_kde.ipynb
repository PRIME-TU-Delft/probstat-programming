# Histogram, Empirical Cumulative Distribution, Kernel Density Estimate

In this section we will be exploring these graphical estimates using simulation. 

## Continuous Data

We will start by sampling from the normal distribution and plotting a histogram, empirical cumulative distribution function (ECDF) and kernel density estimate (KDE).

We will start by comparing a histogram of the sample to the actual distribution. Adjust the sample size to see the effect on the histogram.

We get the probability density function (PDF) of the normal distribution using scipy's `norm.pdf`.

```python
import micropip

await micropip.install("seaborn")
await micropip.install("ipywidgets")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interactive, fixed
from scipy.stats import norm, ecdf, expon, binom
import seaborn as sns
```

```python
def update_histogram(mu_val, scale_val, sample_size):
    rng = np.random.default_rng()
    data_norm = rng.normal(loc=mu_val, scale=scale_val, size=sample_size)
    x = np.linspace(mu_val - 4 * scale_val, mu_val + 4 * scale_val, 1000)
    plt.figure()
    plt.plot(
        x,
        norm.pdf(x, loc=mu_val, scale=scale_val),
        "b-",
        label="Theoretical Normal Distribution",
    )
    plt.hist(data_norm, density=True)
    plt.title(
        f"Histogram of Normal Distribution $\mu$ = {mu_val:.2f}, $\sigma$ = {scale_val:.2f}"
    )
    plt.xlabel("$X$")
    plt.ylabel("Frequency")
    plt.legend()
    plt.show()


mu_true = 1.0
scale_true = 0.5
slider_n = widgets.IntSlider(
    value=50, min=10, max=200, step=10, description="Sample Size"
)
interactive_plot = interactive(
    update_histogram,
    mu_val=fixed(mu_true),
    scale_val=fixed(scale_true),
    sample_size=slider_n,
)
interactive_plot
```

As we can see, the histogram becomes a better estimator for the true distribution when the sample size is larger. A similar effect can be observed in the ECDF. Run the code below to compare the sample ECDF and theoretical ECDF. We can calculate the ECDF of a sample using scipy's `ecdf` and `cdf.evaluate()` functions. To get the CDF of the normal distribution itself, we can call scipy's `norm.cdf` function.

```python
def update_ecdf(mu_val, scale_val, sample_size):
    rng = np.random.default_rng()
    x = np.linspace(mu_val - 4 * scale_val, mu_val + 4 * scale_val, 1000)

    data_norm = rng.normal(loc=mu_val, scale=scale_val, size=sample_size)
    ecdf_data = ecdf(data_norm)
    ecdf_y = ecdf_data.cdf.evaluate(x)

    actual_cdf = norm.cdf(x, loc=mu_val, scale=scale_val)

    plt.figure()
    plt.plot(x, actual_cdf, "b-", linewidth=2, label="Theoretical Normal CDF")

    plt.plot(x, ecdf_y, "r-", label="Sample ECDF")
    plt.title(
        f"ECDF of Normal Distribution $\mu$ = {mu_val:.2f}, $\sigma$ = {scale_val:.2f}"
    )
    plt.xlabel("$X$")
    plt.ylabel("Frequency")
    plt.legend()


mu_true = 1.0
scale_true = 0.5
slider_n = widgets.IntSlider(
    value=50, min=10, max=200, step=10, description="Sample Size"
)
interactive_plot = interactive(
    update_ecdf,
    mu_val=fixed(mu_true),
    scale_val=fixed(scale_true),
    sample_size=slider_n,
)
interactive_plot
```

Finally, we analyze the KDE for the sample and compare it to the theoretical distribution. To do so, we will be using seaborn's `kdeplot` on the sampled data.

Again, feel free to adjust the sample size to see the effects.

```python
def update_kde(mu_val, scale_val, sample_size):
    rng = np.random.default_rng()
    x = np.linspace(mu_val - 4 * scale_val, mu_val + 4 * scale_val, 1000)

    data_norm = rng.normal(loc=mu_val, scale=scale_val, size=sample_size)

    plt.figure()
    plt.hist(data_norm, density=True)
    sns.kdeplot(data_norm, fill=True, label="Sample KDE", color="skyblue", alpha=0.7)
    plt.plot(
        x,
        norm.pdf(x, loc=mu_val, scale=scale_val),
        label="Theoretical Normal Distribution",
    )
    plt.title(
        f"KDE of Normal Distribution $\mu$ = {mu_val:.2f}, $\sigma$ = {scale_val:.2f}"
    )
    plt.xlabel("$X$")
    plt.ylabel("Density")
    plt.legend()


mu_true = 1.0
scale_true = 0.5
slider_n = widgets.IntSlider(
    value=50, min=10, max=200, step=10, description="Sample Size"
)
interactive_plot = interactive(
    update_kde,
    mu_val=fixed(mu_true),
    scale_val=fixed(scale_true),
    sample_size=slider_n,
)
interactive_plot
```

Now let's look at some other distributions. In the code below we will be using the exponential distribution.

```python
def update_exp(lambda_val, sample_size):
    rng = np.random.default_rng()
    x = np.linspace(0, 1 / lambda_val * 5, 1000)

    data_norm = rng.exponential(scale=1 / lambda_val, size=sample_size)
    plt.figure(figsize=(21, 5))

    plt.subplot(1, 3, 1)
    plt.hist(data_norm, density=True)
    plt.plot(
        x,
        expon.pdf(x, scale=1 / lambda_val),
        label="Theoretical Exponential Distribution",
    )
    plt.title(f"Histogram of Exponential Distribution $\lambda$ = {lambda_val:.2f}")
    plt.xlabel("$X$")
    plt.ylabel("Frequency")
    plt.legend()

    plt.subplot(1, 3, 2)
    ecdf_data = ecdf(data_norm)
    ecdf_y = ecdf_data.cdf.evaluate(x)
    actual_cdf = expon.cdf(x, scale=1 / lambda_val)
    plt.plot(x, actual_cdf, "b-", linewidth=2, label="Theoretical Exponential CDF")
    plt.plot(x, ecdf_y, "r-", label="Sample ECDF")
    plt.title(f"ECDF of Exponential Distribution $\lambda$ = {lambda_val:.2f}")
    plt.xlabel("$X$")
    plt.ylabel("Frequency")
    plt.legend()

    plt.subplot(1, 3, 3)
    plt.hist(data_norm, density=True)
    plt.plot(
        x,
        expon.pdf(x, scale=1 / lambda_val),
        label="Theoretical Exponential Distribution",
    )
    sns.kdeplot(data_norm, fill=True, label="Sample KDE", color="skyblue", alpha=0.7)
    plt.title(f"KDE of Exponential Distribution $\lambda$ = {lambda_val:.2f}")
    plt.xlabel("$X$")
    plt.ylabel("Density")
    plt.legend()


lambda_true = 0.5
slider_n = widgets.IntSlider(
    value=50, min=10, max=200, step=10, description="Sample Size"
)
interactive_plot = interactive(
    update_exp,
    lambda_val=fixed(lambda_true),
    sample_size=slider_n,
)
interactive_plot
```

## Discrete Data

In the previous section we looked at the normal and exponential distributions, these being continuous distributions. Now let us take a look at how these graphical estimators would look on a discrete distribution. The distribution we will be looking at is the binomial distribution.

We will start by sampling from the binomial distribution and constructing a histogram based on this sample. Run the code below and adjust the sample size to observe the effect on the resulting histogram. We sample from the binomial distribution using numpy's `random.Generator.binomial` function. To get the probability mass function (PMF) of the distribution we use scipy's `binom.pmf` function.

```python
def update_binomial(n, p, sample_size):
    r_vals = np.arange(n + 1)
    dist_binomial = binom.pmf(r_vals, n, p)
    rng = np.random.default_rng()
    sample = rng.binomial(n, p, size=sample_size)
    freqs = np.bincount(sample.astype(np.int32), minlength=n + 1) / sample_size

    plt.figure(figsize=(8, 5))
    plt.bar(r_vals, freqs, alpha=0.5, label="Sampled Frequency", color="orange")
    plt.plot(r_vals, dist_binomial, "b-", label="Theoretical PMF", linewidth=2)
    plt.xlabel("r")
    plt.ylabel("Probability")
    plt.legend()
    plt.show()


n = 10
p = 0.6
slider_n = widgets.IntSlider(value=5, min=5, max=100, step=5, description="Sample Size")
interactive_plot = interactive(
    update_binomial,
    n=fixed(n),
    p=fixed(p),
    sample_size=slider_n,
)
interactive_plot
```

Next we will look at the ECDF of the binomial distribution. Run the code below to plot the theoretical ECDF and the sample ECDF to compare the two. Adjust the sample size to observe the effect on the sample ECDF.

```python
def update_ecdf_binomial(n, p, sample_size):
    x = np.arange(n + 1)

    data_norm = np.random.binomial(n, p, size=sample_size)
    ecdf_data = ecdf(data_norm)
    ecdf_y = ecdf_data.cdf.evaluate(x)

    actual_cdf = binom.cdf(x, n=n, p=p)

    plt.figure()
    plt.plot(x, actual_cdf, "b-", linewidth=2, label="Theoretical Binomial CDF")

    plt.plot(x, ecdf_y, "r-", label="Sample ECDF")
    plt.title(f"ECDF of Binomial Distribution n = {n}, p = {p:.2f}")
    plt.xlabel("$X$")
    plt.ylabel("Frequency")
    plt.legend()


n = 10
p = 0.6
slider_n = widgets.IntSlider(value=5, min=5, max=100, step=5, description="Sample Size")
interactive_plot = interactive(
    update_ecdf_binomial,
    n=fixed(n),
    p=fixed(p),
    sample_size=slider_n,
)
interactive_plot
```

Finally, we look at the KDE for a sample from the binomial distribution.

```python
def update_binomial(n, p, sample_size):
    r_vals = np.arange(n + 1)
    dist_binomial = binom.pmf(r_vals, n, p)

    rng = np.random.default_rng()
    sample = rng.binomial(n, p, size=sample_size)
    freqs = np.bincount(sample.astype(np.int32), minlength=n + 1) / sample_size

    plt.figure(figsize=(8, 5))
    plt.bar(r_vals, freqs, alpha=0.5, label="Sampled Frequency", color="orange")
    plt.plot(r_vals, dist_binomial, "b-", label="Theoretical PMF", linewidth=2)
    sns.kdeplot(sample, fill=True, label="Sample KDE", color="skyblue", alpha=0.7)
    plt.xlabel("r")
    plt.ylabel("Probability")
    plt.legend()
    plt.show()


n = 10
p = 0.6
slider_n = widgets.IntSlider(value=5, min=5, max=100, step=5, description="Sample Size")
interactive_plot = interactive(
    update_binomial,
    n=fixed(n),
    p=fixed(p),
    sample_size=slider_n,
)
interactive_plot
```
