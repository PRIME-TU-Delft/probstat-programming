# Histogram, Empirical Cumulative Distribution, Kernel Density Estimate

In this section we will be exploring these graphical estimates using simulation. We will start by sampling from the normal distribution and plotting a histogram, empirical cumulative distribution function (ECDF) and kernel density estimate (KDE).

We will start by comparing a histogram of the sample to the actual distribution. Adjust the sample size to see the effect on the histogram.

```python
import micropip

await micropip.install("seaborn")
await micropip.install("ipywidgets")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interactive, fixed
from scipy.stats import norm, ecdf, expon
import seaborn as sns
```

```python
def update_histogram(mu_val, scale_val, sample_size):
    rng = np.random.default_rng()
    data_norm = rng.normal(loc=mu_val, scale=scale_val, size=sample_size)
    x = np.linspace(mu_val - 4 * scale_val, mu_val + 4 * scale_val, 1000)
    plt.figure()
    plt.plot(
        x,
        norm.pdf(x, loc=mu_val, scale=scale_val),
        "b-",
        label="Theoretical Normal Distribution",
    )
    plt.hist(data_norm, density=True)
    plt.title(
        f"Histogram of Normal Distribution $\mu$ = {mu_val:.2f}, $\sigma$ = {scale_val:.2f}"
    )
    plt.xlabel("$X$")
    plt.ylabel("Frequency")
    plt.legend()
    plt.show()


mu_true = 1.0
scale_true = 0.5
slider_n = widgets.IntSlider(
    value=50, min=10, max=200, step=10, description="Sample Size"
)
interactive_plot = interactive(
    update_histogram,
    mu_val=fixed(mu_true),
    scale_val=fixed(scale_true),
    sample_size=slider_n,
)
interactive_plot
```

As we can see, the histogram becomes a better estimator for the true distribution when the sample size is larger. A similar effect can be observed in the ECDF. Run the code below to compare the sample ECDF and theoretical ECDF.

```python
def update_ecdf(mu_val, scale_val, sample_size):
    rng = np.random.default_rng()
    x = np.linspace(mu_val - 4 * scale_val, mu_val + 4 * scale_val, 1000)

    data_norm = rng.normal(loc=mu_val, scale=scale_val, size=sample_size)
    ecdf_data = ecdf(data_norm)
    ecdf_y = ecdf_data.cdf.evaluate(x)

    actual_cdf = norm.cdf(x, loc=mu_val, scale=scale_val)

    plt.figure()
    plt.plot(x, actual_cdf, "b-", linewidth=2, label="Theoretical Normal CDF")

    plt.plot(x, ecdf_y, "r-", label="Sample ECDF")
    plt.title(
        f"ECDF of Normal Distribution $\mu$ = {mu_val:.2f}, $\sigma$ = {scale_val:.2f}"
    )
    plt.xlabel("$X$")
    plt.ylabel("Frequency")
    plt.legend()


mu_true = 1.0
scale_true = 0.5
slider_n = widgets.IntSlider(
    value=50, min=10, max=200, step=10, description="Sample Size"
)
interactive_plot = interactive(
    update_ecdf,
    mu_val=fixed(mu_true),
    scale_val=fixed(scale_true),
    sample_size=slider_n,
)
interactive_plot
```

Finally, we analyze the KDE for the sample and compare it to the theoretical distribution. Again, feel free to adjust the sample size to see the effects.

```python
def update_kde(mu_val, scale_val, sample_size):
    rng = np.random.default_rng()
    x = np.linspace(mu_val - 4 * scale_val, mu_val + 4 * scale_val, 1000)

    data_norm = rng.normal(loc=mu_val, scale=scale_val, size=sample_size)

    plt.figure()
    plt.hist(data_norm, density=True)
    sns.kdeplot(data_norm, fill=True, label="Sample KDE", color="skyblue", alpha=0.7)
    plt.plot(
        x,
        norm.pdf(x, loc=mu_val, scale=scale_val),
        label="Theoretical Normal Distribution",
    )
    plt.title(
        f"KDE of Normal Distribution $\mu$ = {mu_val:.2f}, $\sigma$ = {scale_val:.2f}"
    )
    plt.xlabel("$X$")
    plt.ylabel("Density")
    plt.legend()


mu_true = 1.0
scale_true = 0.5
slider_n = widgets.IntSlider(
    value=50, min=10, max=200, step=10, description="Sample Size"
)
interactive_plot = interactive(
    update_kde,
    mu_val=fixed(mu_true),
    scale_val=fixed(scale_true),
    sample_size=slider_n,
)
interactive_plot
```

Now let's look at some other distributions. In the code below we will be using the exponential distribution.

```python
def update_exp(lambda_val, sample_size):
    rng = np.random.default_rng()
    x = np.linspace(0, 1 / lambda_val * 5, 1000)

    data_norm = rng.exponential(scale=1 / lambda_val, size=sample_size)
    plt.figure(figsize=(21, 5))

    plt.subplot(1, 3, 1)
    plt.hist(data_norm, density=True)
    plt.plot(
        x,
        expon.pdf(x, scale=1 / lambda_val),
        label="Theoretical Exponential Distribution",
    )
    plt.title(f"Histogram of Exponential Distribution $\lambda$ = {lambda_val:.2f}")
    plt.xlabel("$X$")
    plt.ylabel("Frequency")
    plt.legend()

    plt.subplot(1, 3, 2)
    ecdf_data = ecdf(data_norm)
    ecdf_y = ecdf_data.cdf.evaluate(x)
    actual_cdf = expon.cdf(x, scale=1 / lambda_val)
    plt.plot(x, actual_cdf, "b-", linewidth=2, label="Theoretical Exponential CDF")
    plt.plot(x, ecdf_y, "r-", label="Sample ECDF")
    plt.title(f"ECDF of Exponential Distribution $\lambda$ = {lambda_val:.2f}")
    plt.xlabel("$X$")
    plt.ylabel("Frequency")
    plt.legend()

    plt.subplot(1, 3, 3)
    plt.hist(data_norm, density=True)
    plt.plot(
        x,
        expon.pdf(x, scale=1 / lambda_val),
        label="Theoretical Exponential Distribution",
    )
    sns.kdeplot(data_norm, fill=True, label="Sample KDE", color="skyblue", alpha=0.7)
    plt.title(f"KDE of Exponential Distribution $\lambda$ = {lambda_val:.2f}")
    plt.xlabel("$X$")
    plt.ylabel("Density")
    plt.legend()


lambda_true = 0.5
slider_n = widgets.IntSlider(
    value=50, min=10, max=200, step=10, description="Sample Size"
)
interactive_plot = interactive(
    update_exp,
    lambda_val=fixed(lambda_true),
    sample_size=slider_n,
)
interactive_plot
```
