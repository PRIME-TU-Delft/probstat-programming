# Mean Squared Error

In this section we will be exploring mean squared error (MSE). We will be looking at the MSE of a simulated exponential and normal distribution. In order to sample from these distributions, we will use numpy's `random.Generator.exponential` and `random.Generator.normal` functions.

We will start by simulating the exponential distribution to estimate $\lambda$. Run the code blocks below to simulate sampling many times from the exponential distribution. We will use the maximum likelihood estimator to estimate $\lambda$ and plot histograms of the estimates. We also compute the error of the estimates by calculating the difference between each estimate and the true $\lambda$ value. Finally, we calculate the MSE.

```python
import micropip

await micropip.install("ipywidgets")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interactive, fixed
```

```python
lambda_true = 0.5
sample_size = 100
samples = 1000

estimates = []
errors = []

# simulate samples
for i in range(samples):
    rng = np.random.default_rng()
    sample = rng.exponential(scale=1 / lambda_true, size=sample_size)

    mle = sample_size / sample.sum()
    estimates.append(mle)
    error = lambda_true - mle
    errors.append(error)

estimates = np.array(estimates)
mean_estimates = estimates.mean()
errors = np.array(errors)
mse = np.mean(errors**2)

print(f"MSE = {mse:.4f}")

# plot estimates
plt.figure(figsize=(15, 6))
plt.subplot(1, 2, 1)
plt.hist(estimates)
plt.title("Estimation $\hat{\lambda}$")
plt.ylabel("frequency")
plt.xlabel("$\lambda$ estimates")

plt.axvline(x=lambda_true, color="r", linestyle="--")
plt.axvline(x=mean_estimates, color="y", linestyle="--")
plt.legend(["true value", "mean of estimates"])

plt.subplot(1, 2, 2)
plt.hist(errors)
plt.title("Errors MSE")
plt.ylabel("frequency")
plt.xlabel("Error (estimate - true value)")

plt.axvline(x=mse, color="y", linestyle="--")
plt.legend(["mse"])

plt.show()
```

Now lets investigate the impact of $\lambda$ on the MSE. In the code block below we plot the calculated MSE versus the value of $\lambda$. Observe the relationship between $\lambda$ and the MSE.

```python
def simulate_exponential(lambda_vals, sample_size=100, samples=1000):
    mse = []
    # simulate samples
    for i in lambda_vals:
        errors = []
        for j in range(samples):
            rng = np.random.default_rng()
            sample = rng.exponential(scale=1 / i, size=sample_size)

            mle = sample_size / sample.sum()
            error = i - mle
            errors.append(error)

        errors = np.array(errors)
        mse.append(np.mean(errors**2))

    return mse


lambda_vals = np.linspace(0.1, 2, 150)
sample_size = 100
samples = 1000
mse_vals = simulate_exponential(lambda_vals, sample_size, samples)

plt.figure(figsize=(14, 8))


plt.plot(lambda_vals, mse_vals, label="mse values")
plt.title("Mean Squared Error vs Lambda Values")
plt.legend()
plt.xlabel("Lambda Values")
plt.ylabel("MSE")

plt.show()
```

Next, we turn our attention to the normal distribution. 

We will repeat the same thing we did with the exponential distribution but this time to estimate $\mu$ in the normal distribution. Run the code block below to simulate sampling many times from the normal distribution. We will again use the maximum likelihood estimator to estimate $\mu$ and plot histograms of the estimates. We compute the error of the estimates by calculating the difference between each estimate and the true $\mu$ value and calculate the MSE.

```python
mu_true = 2.5
sample_size = 100
samples = 1000

estimates = []
errors = []

# simulate samples
for i in range(samples):
    rng = np.random.default_rng()
    sample = rng.normal(loc=mu_true, scale=1.0, size=sample_size)

    mle = sample.sum() / sample_size
    estimates.append(mle)
    error = mu_true - mle
    errors.append(error)

estimates = np.array(estimates)
mean_estimates = estimates.mean()
errors = np.array(errors)
mse = np.mean(errors**2)

print(f"MSE = {mse:.4f}")

# plot estimates
plt.figure(figsize=(15, 6))
plt.subplot(1, 2, 1)
plt.hist(estimates)
plt.title("Estimation $\hat{\mu}$")
plt.ylabel("frequency")
plt.xlabel("$\mu$ estimates")

plt.axvline(x=mu_true, color="r", linestyle="--")
plt.axvline(x=mean_estimates, color="y", linestyle="--")
plt.legend(["true value", "mean of estimates"])

plt.subplot(1, 2, 2)
plt.hist(errors)
plt.title("Errors MSE")
plt.ylabel("frequency")
plt.xlabel("Error (estimate - true value)")

plt.axvline(x=mse, color="y", linestyle="--")
plt.legend(["mse"])

plt.show()
```

Again, we investigate the impact of the value of $\mu$ on the MSE. Run the code below to plot the MSE versus the value of $\mu$ and observe the relationship between the two values.

```python
def simulate_normal(mu_vals, sample_size=100, samples=1000):
    mse = []
    # simulate samples
    for i in mu_vals:
        errors = []
        for j in range(samples):
            rng = np.random.default_rng()
            sample = rng.normal(loc=i, scale=1.0, size=sample_size)

            mle = sample.sum() / sample_size
            error = i - mle
            errors.append(error)

        errors = np.array(errors)
        mse.append(np.mean(errors**2))

    return mse


mu_vals = np.linspace(0, 2, 150)
sample_size = 100
samples = 1000
mse_vals = simulate_normal(mu_vals, sample_size, samples)

plt.figure(figsize=(14, 8))


plt.plot(lambda_vals, mse_vals, label="mse values")
plt.title("Mean Squared Error vs Mu Values")
plt.legend()
plt.xlabel("Mu Values")
plt.ylabel("MSE")

plt.show()
```

Let us now consider the example where we use the minimum of a sample times the sample size to estimate the expectation of the exponential distribution. Run the code below to plot histograms of the estimates and the associated error values alongside the MSE. Use the slider to adjust the value of $\lambda$.

```python
def update_plot(lambda_value):
    estimates = []
    errors = []

    # simulate samples
    for i in range(samples):
        rng = np.random.default_rng()
        sample = rng.exponential(scale=1.0 / lambda_value, size=sample_size)

        estimate = sample.min() * sample_size
        estimates.append(estimate)
        error = (1.0 / lambda_value) - estimate
        errors.append(error)

    estimates = np.array(estimates)
    mean_estimates = estimates.mean()
    errors = np.array(errors)
    mse = np.mean(errors**2)

    print(f"MSE = {mse:.4f}")

    # plot estimates
    plt.figure(figsize=(15, 6))
    plt.subplot(1, 2, 1)
    plt.hist(estimates)
    plt.title("Estimation $1 / \hat{\lambda}$")
    plt.ylabel("frequency")
    plt.xlabel("$1 / \lambda$ estimates")

    plt.axvline(x=1.0 / lambda_value, color="r", linestyle="--")
    plt.axvline(x=mean_estimates, color="y", linestyle="--")
    plt.legend(["true 1/$\lambda$ value", "mean of estimates"])

    plt.subplot(1, 2, 2)
    plt.hist(errors)
    plt.title("Errors minimum x sample size")
    plt.ylabel("frequency")
    plt.xlabel("Error")

    plt.axvline(x=mse, color="y", linestyle="--")
    plt.legend(["mse"])

    plt.show()


lambda_true = widgets.FloatSlider(
    value=0.5, min=0.1, max=5, step=0.1, description="Lambda"
)
sample_size = 100
samples = 1000
interactive_plot = interactive(update_plot, lambda_value=lambda_true)
interactive_plot
```

Run the code below to plot the MSE versus the value of $\lambda$ to observe the relationship between the two.

```python
def simulate_exponential_bad(lambda_vals, sample_size=100, samples=1000):
    mse = []
    # simulate samples
    for i in lambda_vals:
        errors = []
        for j in range(samples):
            rng = np.random.default_rng()
            sample = rng.exponential(scale=1 / i, size=sample_size)

            mle = sample.min() * sample_size
            error = (1.0 / i) - mle
            errors.append(error)

        errors = np.array(errors)
        mse.append(np.mean(errors**2))

    return mse


lambda_vals = np.linspace(0.1, 2, 150)
sample_size = 100
samples = 1000
mse_vals = simulate_exponential_bad(lambda_vals, sample_size, samples)

plt.figure(figsize=(14, 8))


plt.plot(lambda_vals, mse_vals, label="mse values")
plt.title("Mean Squared Error vs Lambda Values (estimating $1/\lambda$)")
plt.legend()
plt.xlabel("Lambda Values")
plt.ylabel("MSE")

plt.show()
```

As we can observe from these examples, the MSE value is related to the width of the errors histogram. Narrow histograms have a smaller MSE while wider histograms typically have a larger MSE. Additionally, the value of MSE relates to bias. More bias can result in a higher MSE.
