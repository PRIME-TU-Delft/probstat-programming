# Mean Squared Error

In this section we will be exploring mean squared error (MSE). We will be looking at the MSE of a simulated exponential and normal distribution. 

We will start by simulating the exponential distribution to estimate $\lambda$ and plotting histograms of the estimates as well as the errors in the simulation.

```python
import micropip

await micropip.install("ipywidgets")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interactive, fixed
```

```python
lambda_true = 0.5
sample_size = 100
samples = 1000

estimates = []
errors = []

# simulate samples
for i in range(samples):
    sample = np.random.exponential(scale=1 / lambda_true, size=sample_size)

    mle = sample_size / sample.sum()
    estimates.append(mle)
    error = mle - lambda_true
    errors.append(error)

estimates = np.array(estimates)
mean_estimates = estimates.mean()
errors = np.array(errors)
mse = np.mean(errors**2)

print(f"MSE = {mse:.4f}")

# plot estimates
plt.figure(figsize=(15, 6))
plt.subplot(1, 2, 1)
plt.hist(estimates)
plt.title("Estimation $\hat{\lambda}$")
plt.ylabel("frequency")
plt.xlabel("$\lambda$ estimates")

plt.axvline(x=lambda_true, color="r", linestyle="--")
plt.axvline(x=mean_estimates, color="y", linestyle="--")
plt.legend(["true value", "mean of estimates"])

plt.subplot(1, 2, 2)
plt.hist(errors)
plt.title("Errors MSE")
plt.ylabel("frequency")
plt.xlabel("Error (estimate - true value)")

plt.axvline(x=0, color="r", linestyle="--")
plt.axvline(x=np.mean(errors), color="y", linestyle="--")
plt.legend(["zero error", "mean error"])

plt.show()
```

Next, we turn our attention to the normal distribution. 

```python
mu_true = 2.5
sample_size = 100
samples = 1000

estimates = []
errors = []

# simulate samples
for i in range(samples):
    sample = np.random.normal(loc=mu_true, scale=1.0, size=sample_size)

    mle = sample.sum() / sample_size
    estimates.append(mle)
    error = mle - mu_true
    errors.append(error)

estimates = np.array(estimates)
mean_estimates = estimates.mean()
errors = np.array(errors)
mse = np.mean(errors**2)

print(f"MSE = {mse:.4f}")

# plot estimates
plt.figure(figsize=(15, 6))
plt.subplot(1, 2, 1)
plt.hist(estimates)
plt.title("Estimation $\hat{\mu}$")
plt.ylabel("frequency")
plt.xlabel("$\mu$ estimates")

plt.axvline(x=mu_true, color="r", linestyle="--")
plt.axvline(x=mean_estimates, color="y", linestyle="--")
plt.legend(["true value", "mean of estimates"])

plt.subplot(1, 2, 2)
plt.hist(errors)
plt.title("Errors MSE")
plt.ylabel("frequency")
plt.xlabel("Error (estimate - true value)")

plt.axvline(x=0, color="r", linestyle="--")
plt.axvline(x=np.mean(errors), color="y", linestyle="--")
plt.legend(["zero error", "mean error"])

plt.show()
```

Let us now consider the bad example where we use the minimum of a sample times the sample size to estimate the expectation of the exponential distribution. We compare this estimate to that of the maximum likelihood estimator in the code below.

```python
def update_plot_exp(sample_size, lambda_val):
    rng = np.random.default_rng()
    data_exp = rng.exponential(scale=1/lambda_val, size=sample_size)
    min_val = data_exp.min()
    bad_estimate = min_val * sample_size

    exp = np.linspace(0.01, 2.5, 500)

    n = data_exp.size
    sum_x = data_exp.sum()
    logL = -n * np.log(exp) - sum_x / exp


    plt.figure()
    plt.plot(exp, logL, lw=2)
    plt.axvline(sum_x/n, color="C1", ls="--", label=f"MLE $E[X]$={sum_x/n:.2f}")
    plt.axvline(1/lambda_val, color="tab:olive", ls="--", label=f"True value $E[X]$={1/lambda_val:.2f}")
    plt.axvline(bad_estimate, color="tab:green", ls="--", label=f"minimum x sample size $E[X]$={bad_estimate:.2f}")
    plt.title("Exponential Likelihood")
    plt.xlabel("$E[X]$")
    plt.ylabel("$L(E[X])$")
    plt.legend()
    plt.show()

lambda_true = 0.5
slider_n = widgets.IntSlider(
    value=50, min=10, max=200, step=10, description="Sample Size"
)
interactive_plot = interactive(
    update_plot_exp,
    sample_size=slider_n,
  	lambda_val=fixed(lambda_true)
)
interactive_plot
```

```python
lambda_true = 0.5
sample_size = 100
samples = 1000

estimates = []
errors = []
bad_estimates = []
errors_bad = []

# simulate samples
for i in range(samples):
    sample = np.random.exponential(scale=1 / lambda_true, size=sample_size)

    mle = sample_size / sample.sum()
    estimates.append(mle)
    error = mle - lambda_true
    errors.append(error)

    bad_estimate = sample.min() * sample_size
    bad_estimates.append(bad_estimate)
    error_bad = bad_estimate - lambda_true
    errors_bad.append(error_bad)

estimates = np.array(estimates)
mean_estimates = estimates.mean()
errors = np.array(errors)
mse = np.mean(errors**2)
bad_estimates = np.array(bad_estimates)
bad_mean_estimate = bad_estimates.mean()
errors_bad = np.array(errors_bad)
mse_bad = np.mean(errors_bad**2)

print(f"MSE = {mse:.4f}")
print(f"MSE bad = {mse_bad:.4f}")

# plot estimates
plt.figure(figsize=(15, 6))
plt.subplot(1, 2, 1)
plt.hist(errors)
plt.title("Errors MSE")
plt.ylabel("frequency")
plt.xlabel("Error (estimate - true value)")

plt.axvline(x=0, color="r", linestyle="--")
plt.axvline(x=np.mean(mse), color="y", linestyle="--")
plt.legend(["zero error", "mean error"])

plt.subplot(1, 2, 2)
plt.hist(errors_bad)
plt.title("Errors minimum x sample size")
plt.ylabel("frequency")
plt.xlabel("Error (bad estimate - true value)")

plt.axvline(x=0, color="r", linestyle="--")
plt.axvline(x=np.mean(errors_bad), color="y", linestyle="--")
plt.legend(["zero error", "mean error"])

plt.show()
```
