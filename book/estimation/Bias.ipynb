# Bias

In this section we will be looking at how bias plays into estimation. We will be looking at the maximum likelihood estimator (MLE) for $\lambda$ in the exponential distribution and the MLE for the mean in the normal distribution to illustrate this concept.

The MLE for $\lambda$ was derived in [Maximum Likelihood Estimation](Maximum_Likelihood_Estimate.ipynb) and is $\frac{n}{\sum_{i=1}^n x_i}$. This is a biased estimator.

The MLE for $\mu$ in the normal distribution was also derived in [Maximum Likelihood Estimation](Maximum_Likelihood_Estimate.ipynb) and is $\frac 1 n \sum_{i=1}^n x_i$. This is an unbiased estimator.

To illustrate the impact of bias on these estimators, we can simulate the estimates on many samples of data and visualize the results in a histogram.

First, we simulate the exponential distribution.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng(0)
lambda_true = 0.5
sample_size = 100
samples = 1000

estimates = []

# simulate samples
for i in range(samples):
    sample = np.random.exponential(scale=1/lambda_true, size=sample_size)

    mle = sample_size / sample.sum()
    estimates.append(mle)

estimates = np.array(estimates)
mean_estimates = estimates.mean()

print(f"Bias = {mean_estimates - lambda_true:.4f}")

# plot estimates
plt.figure(figsize=(12, 8))
plt.hist(estimates)
plt.title("Estimation $\hat{\lambda}$")
plt.ylabel("frequency")

plt.axvline(x=lambda_true, color="r", linestyle="--")
plt.axvline(x=mean_estimates, color="y", linestyle="--")
plt.legend(["true value", "mean of estimates"])

plt.show()
```

Next we simulate the normal distribution.

```python
rng = np.random.default_rng(0)
mu_true = 0.0
sample_size = 100
samples = 1000

estimates = []

# simulate samples
for i in range(samples):
    sample = rng.normal(loc=mu_true, scale=1.0, size=sample_size)

    mle = sample.sum() / sample_size
    estimates.append(mle)

estimates = np.array(estimates)
mean_estimates = estimates.mean()

print(f"Bias = {mean_estimates - mu_true:.4f}")

# plot estimates
plt.figure(figsize=(12, 8))
plt.hist(estimates)
plt.title("Estimation $\hat{\mu}$")
plt.ylabel("frequency")

plt.axvline(x=mu_true, color="r", linestyle="--")
plt.axvline(x=mean_estimates, color="y", linestyle="--")
plt.legend(["true value", "mean of estimates"])

plt.show()
```
