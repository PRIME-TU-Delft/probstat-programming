# Bias

In this section we will be looking at how bias plays into estimation. We will be looking at the maximum likelihood estimator (MLE) for $\lambda$ in the exponential distribution and the MLE for the mean $\mu$ in the normal distribution to illustrate this concept.

The MLE for $\lambda$ was derived in [Maximum Likelihood Estimation](Maximum_Likelihood_Estimate.ipynb) and is $\frac{n}{\sum_{i=1}^n x_i}$. This is a biased estimator.

The MLE for $\mu$ in the normal distribution was also derived in [Maximum Likelihood Estimation](Maximum_Likelihood_Estimate.ipynb) and is $\frac 1 n \sum_{i=1}^n x_i$. This is an unbiased estimator.

To illustrate the impact of bias on these estimators, we can simulate the estimates on many samples of data and visualise the results in a histogram.

First, we simulate the exponential distribution.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def simulate_exponential(lambda_true=0.5, sample_size=100, samples=1000):
    estimates = []
    running_avg = []

    # simulate samples
    for i in range(samples):
        sample = np.random.exponential(scale=1 / lambda_true, size=sample_size)
        if i == 0:
            running_avg.append(sample_size / sample.sum())
        else:
            running_avg.append(
                (running_avg[-1] * i + sample_size / sample.sum()) / (i + 1)
            )
        mle = sample_size / sample.sum()
        estimates.append(mle)

    est = np.array(estimates)
    mean_est = est.mean()
    return est, mean_est, running_avg
```

```python
lambda_true = 0.5
sample_size = 100
samples = 1000
estimates, mean_estimates, running_avg_lambda = simulate_exponential(
    lambda_true, sample_size, samples
)
print(f"Bias = {mean_estimates - lambda_true:.4f}")

# plot estimates
plt.figure(figsize=(12, 8))
plt.hist(estimates)
plt.title("Estimation of $\hat{\lambda}$")
plt.ylabel("frequency")
plt.xlabel("estimates")

plt.axvline(x=lambda_true, color="r", linestyle="--")
plt.axvline(x=mean_estimates, color="y", linestyle="--")
plt.legend(["true value", "mean of estimates"])

plt.show()
```

Next we simulate the normal distribution.

```python
def simulate_normal(mu_true=4.2, sample_size=100, samples=1000):
    estimates = []
    running_avg = []

    # simulate samples
    for i in range(samples):
        sample = np.random.normal(loc=mu_true, scale=1.0, size=sample_size)
        if i == 0:
            running_avg.append(sample.sum() / sample_size)
        else:
            running_avg.append(
                (running_avg[-1] * i + sample.sum() / sample_size) / (i + 1)
            )
        mle = sample.sum() / sample_size
        estimates.append(mle)

    est = np.array(estimates)
    mean_est = est.mean()
    return est, mean_est, running_avg
```

```python
mu_true = 4.2
sample_size = 100
samples = 1000

estimates, mean_estimates, running_avg_mu = simulate_normal(
    mu_true, sample_size, samples
)

print(f"Bias = {mean_estimates - mu_true:.4f}")

# plot estimates
plt.figure(figsize=(12, 8))
plt.hist(estimates)
plt.title("Estimation of $\hat{\mu}$")
plt.ylabel("frequency")
plt.xlabel("estimates")

plt.axvline(x=mu_true, color="r", linestyle="--")
plt.axvline(x=mean_estimates, color="y", linestyle="--")
plt.legend(["true value", "mean of estimates"])

plt.show()
```

To further illustrate the concept of bias, let us consider a running average of our estimates. 

The code below displays a running average of the estimates calculated for $\lambda$ in the exponential distribution and $\mu$ in the normal distribution. Feel free to run the code multiple times to observe the variability in behavior.

```python
lambda_true = 0.5
mu_true = 4.2
sample_size = 100
samples = 1000
_, _, running_avg_lambda = simulate_exponential(lambda_true, sample_size, samples)
_, _, running_avg_mu = simulate_normal(mu_true, sample_size, samples)

x = np.arange(1, samples + 1)
figure, axis = plt.subplots(2, 1, figsize=(14, 15))
axis[0].plot(x, running_avg_lambda, label="running average")
axis[0].axhline(y=lambda_true, color="r", linestyle="--", label="true value")
axis[0].set_title("Running Average of Estimates for $\lambda$")
axis[0].legend()
axis[0].set_xlabel("Sample Index")
axis[0].set_ylabel("Estimate Value")

axis[1].plot(x, running_avg_mu, label="running average")
axis[1].axhline(y=mu_true, color="r", linestyle="--", label="true value")
axis[1].set_title("Running Average of Estimates for $\mu$")
axis[1].legend()
axis[1].set_xlabel("Sample Index")
axis[1].set_ylabel("Estimate Value")

plt.show()
```

Initially the behavior of the running average is erratic, but converges in the long run. The dashed line in the plots is the true parameter. If the running average converges to a point above the dashed line, then there is positive bias. If the running average converges to a point below the dashed line, then there is negative bias. This example illustrates the positive bias in our estimator for $\lambda$ and that the estimator for $\mu$ is unbiased.
