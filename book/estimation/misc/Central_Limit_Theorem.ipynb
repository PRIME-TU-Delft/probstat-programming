## Central Limit Theorem (for sums)

The CLT for sums stipulates that for a sequence $X_i,...,X_n$ of independent and identically distributed random variables with expectation $E[X]=\mu$ and variance $Var(X)=\sigma ^ 2$ with large enough $n$, $S_n=\sum^n_{i=1}X_i$ approximately has a normal distribution.

$$
S_n \sim N(n\mu,n\sigma ^ 2)
$$

To illustrate the CLT in action, let us consider die throws. 

```python
def dice_sum_simulation(n, n_simulations):
    dice_rolls = np.random.randint(1, 7, size=(n_simulations, n))
    sums = np.sum(dice_rolls, axis=1)
    return sums


def update_plot(n_dice, n_sim):
    mu_die = 3.5
    var_die = 35 / 12

    mu_sum = n_dice * mu_die
    var_sum = n_dice * var_die
    sigma_sum = np.sqrt(var_sum)

    sums = dice_sum_simulation(n_dice, n_sim)

    plt.figure(figsize=(12, 8))
    plt.hist(
        sums,
        bins=50,
        density=True,
        alpha=0.7,
        color="skyblue",
        edgecolor="black",
        label=f"Simulated Sums (n={n_sim})",
    )
    x = np.linspace(sums.min(), sums.max(), 1000)
    normal_pdf = ss.norm.pdf(x, loc=mu_sum, scale=sigma_sum)
    plt.plot(
        x,
        normal_pdf,
        "r-",
        linewidth=3,
        label=f"Normal Approx: N({mu_sum:.1f}, {sigma_sum:.2f}²)",
    )

    plt.xlabel("Sum of Dice")
    plt.ylabel("Density/Probability")
    plt.title(f"Distribution of Sum of {n_dice} Dice Throws")
    plt.legend()
    plt.grid(True, alpha=0.3)


amt_simulations = 10000

slider_n = widgets.IntSlider(value=2, min=2, max=150, step=2, description="Num dice")
interactive_plot = interactive(
    update_plot, n_dice=slider_n, n_sim=fixed(amt_simulations)
)
interactive_plot
```

## Central Limit Theorem (for averages)

The CLT for averages stipulates that for a sequence $X_1,...,X_n$ of independent and identically distributed random variables with expectation $E[X]=\mu$ and variance $Var(X)=\sigma ^2$ with large enough $n$, $\bar{X}_n = \frac 1 n \sum^n_{i=1} X_i$ approximately has a normal distribution.

$$
\bar{X}_n \sim N(\mu, \frac{\sigma ^ 2}{n})
$$

To illustrate the CLT in action, let us consider an $Exp(1)$ distribution. We will plot the exact distribution and the CLT approximation with various $n$ values.

```python
def update_plots(lambda_true, n):
    x = np.linspace(0, 4.5, 1000)

    pdf_exact = ss.gamma.pdf(x, a=n, scale=1 / (n * lambda_true))
    pdf_normal = ss.norm.pdf(
        x, loc=1 / lambda_true, scale=1 / (lambda_true * np.sqrt(n))
    )

    plt.figure(figsize=(12, 9))

    plt.plot(
        x,
        pdf_normal,
        label=f"N({(1/lambda_true):.2f},{(1/(lambda_true * np.sqrt(n))):.2f})",
        color="tab:orange",
    )
    plt.plot(
        x, pdf_exact, label=f"Mean of {n} i.i.d Exp({lambda_true})", color="tab:blue"
    )
    plt.legend()
    plt.show()


lambda_val = 1.0

slider_n = widgets.IntSlider(value=2, min=2, max=100, step=2, description="Size of n")
interactive_plot = interactive(update_plots, n=slider_n, lambda_true=fixed(lambda_val))
interactive_plot
```
