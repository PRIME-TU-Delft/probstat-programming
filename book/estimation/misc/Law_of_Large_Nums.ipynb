# Law of Large Numbers

This section we will be looking at the law of large numbers. This theorem states that for an independent and identically distributed sequence $X_1,...,X_n$ with $E[X_i]=\mu$ and $E[X_i^2]<\infty$ for any $\epsilon > 0$ the following holds:

$$
P(|\bar{X}_n-\mu|>\epsilon) \rightarrow 0, \text{ as } n \rightarrow \infty
$$

We can illustrate this theorem by considering a simulation of throwing two dice. We will simulate the mean of the maximum value of both dice and the mean of the sum of both dice over a large number of throws.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as ss
```

```python
running_avg_max = []
running_avg_sum = []
throws = 1000

for i in range(throws):
    d1 = np.random.uniform(1,7,1).astype(int)
    d2 = np.random.uniform(1,7,1).astype(int)

    if i==0:
        running_avg_max.append(max(d1,d2))
        running_avg_sum.append(d1 + d2)
    else:
        running_avg_max.append(
                (running_avg_max[-1] * i + max(d1,d2)) / (i + 1)
            )
        running_avg_sum.append(
                (running_avg_sum[-1] * i + (d1+d2)) / (i + 1)
        )

x = np.arange(1, throws + 1)
figure, axis = plt.subplots(2, 1, figsize=(14, 15))
axis[0].plot(x, running_avg_max, label="running average")
axis[0].axhline(y=4.5, color="r", linestyle="--", label="true value")
axis[0].set_title("Mean of max of two dice")
axis[0].legend()
axis[0].set_xlabel("Throws")
axis[0].set_ylabel("Mean number")

axis[1].plot(x, running_avg_sum, label="running average")
axis[1].axhline(y=7, color="r", linestyle="--", label="true value")
axis[1].set_title("Mean of sum of two dice")
axis[1].legend()
axis[1].set_xlabel("Throws")
axis[1].set_ylabel("Mean number")

plt.show()
```

As this simulation illustrates, with large $n$, the average of the trial results converges to the expected value.
