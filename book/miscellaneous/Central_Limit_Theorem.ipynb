## Central Limit Theorem (for sums)

The CLT for sums stipulates that for a sequence $X_i,...,X_n$ of independent and identically distributed random variables with expectation $E[X]=\mu$ and variance $Var(X)=\sigma ^ 2$ with large enough $n$, $S_n=\sum^n_{i=1}X_i$ approximately has a normal distribution.

$$
S_n \sim N(n\mu,n\sigma ^ 2)
$$

To illustrate the CLT in action, let us consider several different distributions. Run the code below to show the distribution of the sum of independent samples from the exponential distribution.

```python
def update_plot(sample_size, n_sim, scale):
    rng = np.random.default_rng()
    sample = rng.exponential(scale=scale, size=(n_sim, sample_size))
    sums = np.sum(sample, axis=1)

    expectation_exp = scale
    var_exp = scale**2

    mu_sum = sample_size * expectation_exp
    var_sum = sample_size * var_exp
    sigma_sum = np.sqrt(var_sum)

    plt.figure(figsize=(12, 8))
    plt.hist(
        sums,
        bins=50,
        alpha=0.7,
        density=True,
        color="skyblue",
        edgecolor="black",
        label=f"Simulated Sums (n={n_sim})",
    )
    x = np.linspace(sums.min(), sums.max(), 1000)
    normal_pdf = ss.norm.pdf(x, loc=mu_sum, scale=sigma_sum)
    plt.plot(
        x,
        normal_pdf,
        "r-",
        linewidth=3,
        label=f"Normal Approx: N({mu_sum:.1f}, {var_sum:.2f})",
    )

    plt.xlabel(f"Sum of Exponential Distribution Samples (lambda={1/scale})")
    plt.ylabel("Density/Probability")
    plt.title(
        f"Distribution of Sum of Samples n={sample_size} from Exponential Distribution"
    )
    plt.legend()
    plt.grid(True, alpha=0.3)


scale_slider = widgets.FloatSlider(
    value=2.0, min=0.1, max=5.0, step=0.1, description="scale"
)
slider_n = widgets.IntSlider(value=2, min=2, max=150, step=2, description="Sample Size")
slider_sim = widgets.IntSlider(
    value=1000, min=1000, max=10000, step=500, description="Number of Simulations"
)
interactive_plot = interactive(
    update_plot, sample_size=slider_n, n_sim=slider_sim, scale=scale_slider
)
interactive_plot
```

Now let us consider the uniform distribution.

```python
def update_plot(sample_size, n_sim, range_vals):
    a, b = range_vals

    rng = np.random.default_rng()
    sample = rng.uniform(a, b, size=(n_sim, sample_size))
    sums = np.sum(sample, axis=1)

    expectation_uniform = (a + b) / 2
    var_uniform = (b - a) ** 2 / 12

    mu_sum = sample_size * expectation_uniform
    var_sum = sample_size * var_uniform
    sigma_sum = np.sqrt(var_sum)

    plt.figure(figsize=(12, 8))
    plt.hist(
        sums,
        bins=50,
        alpha=0.7,
        density=True,
        color="skyblue",
        edgecolor="black",
        label=f"Simulated Sums (n={n_sim})",
    )
    x = np.linspace(sums.min(), sums.max(), 1000)
    normal_pdf = ss.norm.pdf(x, loc=mu_sum, scale=sigma_sum)
    plt.plot(
        x,
        normal_pdf,
        "r-",
        linewidth=3,
        label=f"Normal Approx: N({mu_sum:.1f}, {var_sum:.2f})",
    )

    plt.xlabel(f"Sum of Uniform Distribution Samples a={a}, b={b}")
    plt.ylabel("Density/Probability")
    plt.title(
        f"Distribution of Sum of Samples n={sample_size} from Uniform Distribution"
    )
    plt.legend()
    plt.grid(True, alpha=0.3)


slider_range = widgets.IntRangeSlider(
    values=[1, 2], min=-1, max=5, step=0.5, description="a,b"
)
slider_n = widgets.IntSlider(value=2, min=2, max=150, step=2, description="Sample Size")
slider_sim = widgets.IntSlider(
    value=100, min=100, max=10000, step=500, description="Number of Simulations"
)
interactive_plot = interactive(
    update_plot, sample_size=slider_n, n_sim=slider_sim, range_vals=slider_range
)
interactive_plot
```

Now we can look at the geometric distribution.

```python
def update_plot(sample_size, n_sim, p):

    rng = np.random.default_rng()
    sample = rng.geometric(p, size=(n_sim, sample_size))
    sums = np.sum(sample, axis=1)

    expectation_geom = 1 / p
    var_geom = (1 - p) / p**2

    mu_sum = sample_size * expectation_geom
    var_sum = sample_size * var_geom
    sigma_sum = np.sqrt(var_sum)

    plt.figure(figsize=(12, 8))
    plt.hist(
        sums,
        bins=50,
        alpha=0.7,
        density=True,
        color="skyblue",
        edgecolor="black",
        label=f"Simulated Sums (n={n_sim})",
    )
    x = np.linspace(sums.min(), sums.max(), 1000)
    normal_pdf = ss.norm.pdf(x, loc=mu_sum, scale=sigma_sum)
    plt.plot(
        x,
        normal_pdf,
        "r-",
        linewidth=3,
        label=f"Normal Approx: N({mu_sum:.1f}, {var_sum:.2f})",
    )

    plt.xlabel(f"Sum of Geometric Distribution Samples p={p}")
    plt.ylabel("Density/Probability")
    plt.title(
        f"Distribution of Sum of Samples n={sample_size} from Geometric Distribution"
    )
    plt.legend()
    plt.grid(True, alpha=0.3)


p_slider = widgets.FloatSlider(value=0.5, min=0.1, max=1.0, step=0.1, description="p")
slider_n = widgets.IntSlider(value=2, min=2, max=150, step=2, description="Sample Size")
slider_sim = widgets.IntSlider(
    value=100, min=100, max=10000, step=500, description="Number of Simulations"
)
interactive_plot = interactive(
    update_plot, sample_size=slider_n, n_sim=slider_sim, p=p_slider
)
interactive_plot
```

Let us also consider a bimodal distribution.

```python
def update_plot(mu1, mu2, n, samples):
    rng = np.random.default_rng()
    faces = 3
    s1 = 1.0
    s2 = 1.0

    w1 = faces / 6.0
    w2 = 1.0 - w1
    mu = w1 * mu1 + w2 * mu2
    var = w1 * (s1**2 + mu1**2) + w2 * (s2**2 + mu2**2) - mu**2
    sigma = math.sqrt(var) if var > 0 else 0.0

    faces_draws = rng.integers(1, 7, size=(samples, n))
    data = np.empty_like(faces_draws, dtype=float)

    mask1 = faces_draws <= faces
    mask2 = ~mask1

    if mask1.any():
        data[mask1] = rng.normal(loc=mu1, scale=s1, size=mask1.sum())
    if mask2.any():
        data[mask2] = rng.normal(loc=mu2, scale=s2, size=mask2.sum())

    sums = data.sum(axis=1)
    means = sums / n
    std_means = (
        (means - mu) / (sigma / math.sqrt(n)) if sigma > 0 else np.zeros_like(means)
    )

    mu_sum = n * mu
    sigma_sum = math.sqrt(n * var) if var > 0 else 0.0
    mu_mean = mu
    sigma_mean = math.sqrt(var / n) if var > 0 else 0.0

    plt.figure(figsize=(12, 8))
    plt.hist(sums, bins=50, density=True, alpha=0.6)
    x = np.linspace(sums.min(), sums.max(), 300)
    if sigma_sum > 0:
        plt.plot(
            x,
            norm.pdf(x, loc=mu_sum, scale=sigma_sum),
            lw=2,
            label=f"N({mu_sum:.2f}, {sigma_sum**2:.2f})",
        )
    plt.title(f"Sum of {n} draws\n(faces {faces}/6 → comp1)")
    plt.legend()
    plt.grid(True, alpha=0.3)


mu1_slider = widgets.FloatSlider(
    value=-2.0, min=-6.0, max=6.0, step=0.1, description="mu1"
)
mu2_slider = widgets.FloatSlider(
    value=2.0, min=-6.0, max=6.0, step=0.1, description="mu2"
)
n_slider = widgets.IntSlider(value=5, min=1, max=200, step=1, description="n")
slider_sim = widgets.IntSlider(
    value=100, min=100, max=10000, step=500, description="Number of Simulations"
)

interactive_plot = interactive(
    update_plot,
    mu1=mu1_slider,
    mu2=mu2_slider,
    n=n_slider,
    samples=slider_sim,
)

interactive_plot
```

## Central Limit Theorem (for averages)

The CLT for averages stipulates that for a sequence $X_1,...,X_n$ of independent and identically distributed random variables with expectation $E[X]=\mu$ and variance $Var(X)=\sigma ^2$ with large enough $n$, $\bar{X}_n = \frac 1 n \sum^n_{i=1} X_i$ approximately has a normal distribution.

$$
\bar{X}_n \sim N(\mu, \frac{\sigma ^ 2}{n})
$$

To illustrate the CLT in action, let us consider the same distributions that we did for the CLT for sums. First we will plot the averages and then we will plot the standardized averages. We start with an exponential distribution. Feel free to adjust the parameter values using the sliders.

```python
def update_plot(sample_size, n_sim, scale):
    rng = np.random.default_rng()
    sample = rng.exponential(scale=scale, size=(n_sim, sample_size))
    avgs = np.mean(sample, axis=1)

    expectation_exp = scale
    var_exp = scale**2

    mu_avg = expectation_exp
    var_avg = var_exp / sample_size
    sigma_avg = np.sqrt(var_avg)

    plt.figure(figsize=(12, 8))
    plt.subplot(2, 1, 1)

    plt.hist(
        avgs,
        bins=50,
        alpha=0.7,
        density=True,
        color="skyblue",
        edgecolor="black",
        label=f"Simulated Averages (n={n_sim})",
    )
    x = np.linspace(avgs.min(), avgs.max(), 1000)
    normal_pdf = ss.norm.pdf(x, loc=mu_avg, scale=sigma_avg)
    plt.plot(
        x,
        normal_pdf,
        "r-",
        linewidth=3,
        label=f"Normal Approx: N({mu_avg:.1f}, {var_avg:.2f})",
    )

    plt.xlabel(f"Averages of Exponential Distribution Samples (lambda={(1/scale):2f})")
    plt.ylabel("Density/Probability")
    plt.title(
        f"Distribution of Averages of Samples n={sample_size} from Exponential Distribution"
    )
    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.subplot(2, 1, 2)
    avg_standard = (avgs - expectation_exp) / (np.sqrt(var_exp / sample_size))
    plt.hist(
        avg_standard,
        bins=50,
        alpha=0.7,
        density=True,
        color="skyblue",
        edgecolor="black",
        label=f"Simulated Standardized Averages (n={n_sim})",
    )
    x = np.linspace(avg_standard.min(), avg_standard.max(), 1000)
    normal_pdf = ss.norm.pdf(x, loc=0.0, scale=1.0)
    plt.plot(
        x,
        normal_pdf,
        "r-",
        linewidth=3,
        label=f"Normal Approx: N({0:.1f}, {1:.2f})",
    )
    plt.xlabel(
        f"Standardized Averages of Exponential Distribution Samples (lambda={(1/scale):2f})"
    )
    plt.ylabel("Density/Probability")
    plt.title(
        f"Distribution of Standardized Averages of Samples n={sample_size} from Exponential Distribution"
    )
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()


scale_slider = widgets.FloatSlider(
    value=2.0, min=0.1, max=5.0, step=0.1, description="scale"
)
slider_n = widgets.IntSlider(value=2, min=2, max=150, step=2, description="Sample Size")
slider_sim = widgets.IntSlider(
    value=1000, min=1000, max=10000, step=500, description="Number of Simulations"
)
interactive_plot = interactive(
    update_plot, sample_size=slider_n, n_sim=slider_sim, scale=scale_slider
)
interactive_plot
```
