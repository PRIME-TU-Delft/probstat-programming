# Law of Large Numbers

This section we will be looking at the law of large numbers. This theorem states that for an independent and identically distributed sequence $X_1,...,X_n$ with $E[X_i]=\mu$ and $E[X_i^2]<\infty$ for any $\epsilon > 0$ the following holds:

$$
P(|\bar{X}_n-\mu|>\epsilon) \rightarrow 0, \text{ as } n \rightarrow \infty
$$

We can illustrate this theorem by considering simulations of different distribution types. We will begin by simulating the uniform distribution. Run the code below and adjust the parameter to observe the effect.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as ss
import ipywidgets as widgets
from ipywidgets import interactive, fixed
```

```python
def update_plot(range_vals, samples):
    a, b = range_vals
    running_avg = []
    for i in range(samples):
        rng = np.random.default_rng()
        u1 = rng.uniform(a, b)

        if i == 0:
            running_avg.append(u1)
        else:
            running_avg.append((running_avg[-1] * i + u1) / (i + 1))

    x = np.arange(1, samples + 1)
    plt.figure(figsize=(14, 8))
    plt.plot(x, running_avg, label="running average")
    plt.axhline(y=(a + b) / 2, color="r", linestyle="--", label="true value")
    plt.title(f"Running average of uniform distribution alpha={a:.1f}, beta={b:.1f}")
    plt.legend()
    plt.xlabel("n")
    plt.ylabel("Running average")

    plt.show()


samples = 1000
slider_a = widgets.IntRangeSlider(
    values=[1, 2], min=-1, max=5, step=0.5, description="a,b"
)
interactive_plot = interactive(update_plot, range_vals=slider_a, samples=fixed(samples))
interactive_plot
```

As this simulation illustrates, with large $n$, the average of the values sampled from the uniform distribution converges to the expected value.

Now let us consider a different distribution. Run the code below to plot the running average of the exponential distribution.

```python
def update_plot(lambda_val, samples):
    running_avg = []
    for i in range(samples):
        rng = np.random.default_rng()
        e1 = rng.exponential(lambda_val)

        if i == 0:
            running_avg.append(e1)
        else:
            running_avg.append((running_avg[-1] * i + e1) / (i + 1))

    x = np.arange(1, samples + 1)
    plt.figure(figsize=(14, 8))
    plt.plot(x, running_avg, label="running average")
    plt.axhline(y=1 / lambda_val, color="r", linestyle="--", label="true value")
    plt.title(f"Running average of exponential distribution lambda={lambda_val:.2f}")
    plt.legend()
    plt.xlabel("n")
    plt.ylabel("Running average")

    plt.show()


samples = 1000
slider_lambda = widgets.FloatSlider(
    value=1, min=0.1, max=5, step=0.1, description="Lambda"
)
interactive_plot = interactive(
    update_plot, lambda_val=slider_lambda, samples=fixed(samples)
)
interactive_plot
```

Finally, let's look at the Pareto distribution. Experiment with values of $\alpha$ that are below $1$. Observe how the running average does not converge to the expected average when this is the case.

```python
def update_plot(a, samples):
    running_avg = []
    for i in range(samples):
        rng = np.random.default_rng()
        p = rng.pareto(a)

        if i == 0:
            running_avg.append(p)
        else:
            running_avg.append((running_avg[-1] * i + p) / (i + 1))

    x = np.arange(1, samples + 1)
    plt.figure(figsize=(14, 8))
    plt.plot(x, running_avg, label="running average")
    plt.axhline(y=1 / (a - 1), color="r", linestyle="--", label="true value")
    plt.title(f"Running average of Pareto distribution alpha={a}")
    plt.legend()
    plt.xlabel("n")
    plt.ylabel("Running average")

    plt.show()


samples = 1000
slider_a = widgets.FloatSlider(value=1.1, min=0.5, max=5, step=0.1, description="a")
interactive_plot = interactive(update_plot, a=slider_a, samples=fixed(samples))
interactive_plot
```
