# Multiple Linear Regression

In this section we will be looking at multiple linear regression. That is, we will use multiple variables to try to predict another variable. We will be using a ford focus dataset and we want a model for the price explained by the mileage, engine size, and age. 

First we will calculate the estimated regression line using statsmodel.

````python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import statsmodels.api as sm
````

````python
data = pd.read_csv("data/Focus.csv")

# Set up observation vector and design matrix
X = data[["mileage", "engineSize", "age"]]
y = data["price"]
X = sm.add_constant(X)

# Fit the model
est = sm.OLS(y, X).fit()

# Print results
print(est.summary())
est.params
````

```{note}
The coefficients can be found under the `coeff` column of the summary (or by looking at `est.params`), number of observations is marked by `No. Observations` and degrees of freedom can be found at `Df Model`.
```

Now we will plot the scatterplot against each of the three explanatory variables. 

````python
# Extracting variables for plotting
mileages = data["mileage"]
engine_sizes = data["engineSize"]
ages = data["age"]
prices = data["price"]

# Plotting the scatterplot against each of the three explanatory variables
plt.figure(figsize=(12, 16))
plt.subplot(3, 1, 1)
plt.scatter(mileages, prices, label="Data Points", color="blue")
plt.xlabel("Mileage")
plt.ylabel("Price")
plt.title("Price vs Mileage")
plt.legend()
plt.grid()

plt.subplot(3, 1, 2)
plt.scatter(engine_sizes, prices, label="Data Points", color="blue")
plt.xlabel("Engine Size")
plt.ylabel("Price")
plt.title("Price vs Engine Size")
plt.legend()
plt.grid()

plt.subplot(3, 1, 3)
plt.scatter(ages, prices, label="Data Points", color="blue")
plt.xlabel("Age")
plt.ylabel("Price")
plt.title("Price vs Age")
plt.legend()
plt.grid()

plt.show()
````

Plotting the estimated regression line would require providing a value for the other two variables. We can remedy this issue by plotting a scatterplot of the response against the value predicted by our model and plotting a slope $1$ line through the origin.

````python
# Calculate x and y values for the slope 1 line
x_vals = np.linspace(-5000, 20000)
y_vals = 1 * x_vals

# Plotting the predicted vs actual price
plt.figure(figsize=(10, 6))
plt.scatter(est.predict(X), prices, label="Data Points", color="blue")
plt.plot(x_vals, y_vals, color="red")
plt.xlabel("Predicted Price")
plt.ylabel("Actual Price")
plt.title("Multiple Linear Regression: Predicted vs Actual Price")
plt.legend()
plt.grid()
plt.show()
````
