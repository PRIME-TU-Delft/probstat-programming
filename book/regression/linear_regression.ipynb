# Simple Linear Regression Example

In this section we will be covering an example of simple linear regression using data. We will be using Ford Focus data to create a model for the price explained by the mileage. 

The first step in this process is to wrangle our data. Run the code below to extract prices and mileage from the dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

```python
data = pd.read_csv("data/Focus.csv")

# Extracting variables for regression
prices = data["price"]
mileages = data["mileage"]

prices, mileages
```

Now we will set up the observation vector and design matrix and use statsmodel to calculate the regression line. We can read off the estimated coefficients throught `results.params` or in the `coef` column of the results summary. The number of observations is listed next to the row labelled `No. Observations` and degrees of freedom is found at `Df Model`.

```python
# Set up observation vector and design matrix
Y = prices
X = sm.add_constant(mileages)

# Fit the OLS model
model = sm.OLS(Y, X)
results = model.fit()

# Return the summary of the regression results
print(results.summary())
results.params
```

Now we can plot the scatterplot and our estimated regression line.

```python
plt.figure(figsize=(10, 6))
# Plotting the scatterplot of mileage vs price
plt.scatter(mileages, prices, label="Data Points", color="blue")
# Plotting the regression line
plt.plot(mileages, results.predict(X), color="red", label="Regression Line")
plt.xlabel("Mileage")
plt.ylabel("Price")
plt.title("Simple Linear Regression: Price vs Mileage")
plt.legend()
plt.grid()
plt.show()
```
