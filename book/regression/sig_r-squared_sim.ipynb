# Significance and R-squared Through Simulation

In this section we will be simulating $Y=a + bX +U$ where $X$ is uniformly distributed and $U$ is normally distributed. We will be adapting each variable to observe the effect on the significance (p-value) and $R^2$.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import ipywidgets as widgets
from ipywidgets import interactive
import statsmodels.api as sm
```

```python
def update_plot(a, b, variance, sample_size):
    rng = np.random.default_rng()
    X = rng.uniform(0, 45, sample_size)
    U = rng.normal(0, np.sqrt(variance), sample_size)
    Y = a + b * X + U

    y = Y
    X = sm.add_constant(X)

    # Fit the model
    est = sm.OLS(y, X).fit()

    # Print results
    print(est.summary())

    values = Y
    estimated_values = est.predict(X)
    # Calculate x and y values for the slope 1 line
    x_vals = np.linspace(np.min(estimated_values), np.max(estimated_values))
    y_vals = 1 * x_vals

    # Plotting the predicted vs actual values
    plt.figure(figsize=(10, 6))
    plt.scatter(estimated_values, values, label="Data Points", color="blue")
    plt.plot(x_vals, y_vals, color="red")
    plt.xlabel("Predicted Value")
    plt.ylabel("Actual Value")
    plt.title("Multiple Linear Regression: Predicted vs Actual values")
    plt.legend()
    plt.grid()
    plt.show()


slider_a = widgets.FloatSlider(
    value=0, min=-100, max=100, step=1, description="Intercept (a)"
)
slider_b = widgets.FloatSlider(
    value=0.1, min=-10, max=10, step=0.1, description="Slope (b)"
)
slider_var = widgets.IntSlider(
    value=20, min=0, max=1000, step=1, description="Variance"
)
slider_sample_size = widgets.IntSlider(
    value=500, min=100, max=1000, step=50, description="Sample Size"
)

interactive_plot = interactive(
    update_plot,
    a=slider_a,
    b=slider_b,
    variance=slider_var,
    sample_size=slider_sample_size,
)
interactive_plot
```

Use the sliders above to adjust the values of each variable and observe how this impacts the significance and $R^2$. Notice how the significance depends on a ratio of the $b$ value to the variance of $U$.
