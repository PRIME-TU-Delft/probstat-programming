# Dependent Explanatory Variables

In this section we will be exploring dependent explanatory variables. We will be using a datset consisting of anonymized CSE1200 Calculus scores consisting of midterm, endterm, final, and resit grade. We will be trying to construct a linear model for the resit grade explained by the midterm, endterm, and final grade.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import statsmodels.api as sm
```

```python
data = data = pd.read_csv("data/resit_final_endterm_midterm.csv")

# Set up observation vector and design matrix
X = data[["Midterm_Grade", "Final_grade", "Endterm_Grade"]]
y = data["Resit_grade"]
X = sm.add_constant(X)

# Fit the model
est = sm.OLS(y, X).fit()

# Print results
print(est.summary())
est.params
```

Now we plot the predicted versus actual grade values. Notice the large values of the coefficients and the large p-value for each coefficient against the performance in the plot. Also note the low p-value of the F-test. All of these observations are indicators that there is a problem with dependence of random variables.

```python
# Calculate x and y values for the slope 1 line
x_vals = np.linspace(0, 10)
y_vals = 1 * x_vals
grades = data["Resit_grade"]

# Plotting the predicted vs actual grade
plt.figure(figsize=(10, 6))
plt.scatter(est.predict(X), grades, label="Data Points", color="blue")
plt.plot(x_vals, y_vals, color="red")
plt.xlabel("Predicted Grade")
plt.ylabel("Actual Grade")
plt.title("Multiple Linear Regression: Predicted vs Actual Grade")
plt.legend()
plt.grid()
plt.show()
```

Indeed it is the case that we have a dependent explanatory variable, the final grade is a weighted average of the midterm and endterm grades. The high p-values for each variable mean that the explanatory variable does not add to the model if the other two are present. This is the case for all of the variables as all information is contained in the other two variables already. The low F-test value and above plot indicate that there is some explanatory value in the variables themselves. The high coefficient values are due to rounding errors since at some point there is a division by the determinant of $X^T X$, which is $0$, but in this case very small due to rounding errors.

Now let us run the model again using only two of the three variables. Feel free to switch around which variables are used to ensure that it does not matter which two are selected. Notice that we achieve the same performance, but now the variables have lower p-values.

```python
# Set up observation vector and design matrix again for the second model, feel free to use any two of the three variables
X = data[["Midterm_Grade", "Endterm_Grade"]]
y = data["Resit_grade"]
X = sm.add_constant(X)

# Fit the model
est = sm.OLS(y, X).fit()

# Print results
print(est.summary())
est.params
```
