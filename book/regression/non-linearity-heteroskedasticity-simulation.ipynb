# Non-linearity, Hetero-skedasticity, and Normality Through Simulation

In this section we will be using simulation to explore non-linearity, hetero-skedasticity and normality. We will begin by simulating $Y=a+bX+cX^2 + U$. $X$ will be uniformly distributed and $U$ will be shifted exponentially distributed. Run the code below to run the simple linear regression model on ($X,Y$) and plot the regression line in a scatterplot of $X$ vs $Y$. Note the p-values and $R^2$ values given in the summary.

````python
import matplotlib.pyplot as plt
import numpy as np
import statsmodels.api as sm
````

````python
# Simulate non-linearity
sample_size = 500
rng = np.random.default_rng()
X = rng.uniform(0, 5, sample_size)
U = rng.exponential(1, sample_size) - 1.0
a = 2
b = 2
c = 3
Y = a + b * X + c * X**2 + U

y = Y
X_cons = sm.add_constant(X)

# Fit the model
est = sm.OLS(y, X_cons).fit()

# Print results
print(est.summary())

values = Y
estimated_values = est.predict(X_cons)

plt.figure(figsize=(10, 6))
plt.scatter(X, values, label="Data Points", color="blue")
plt.plot(X, estimated_values, color="red", label="Regression Line")
plt.xlabel("X Value")
plt.ylabel("Y Value")
plt.title("Linear Regression: X vs Y")
plt.legend()
plt.grid()
plt.show()
````

Now we will do residual analysis. Run the code block below to plot $X$ against the residuals as well as a histogram of the residuals.

````python
residuals = est.resid

plt.figure(figsize=(10, 6))
plt.subplot(2, 1, 1)
plt.scatter(X, residuals, alpha=0.6, color="blue")
plt.axhline(y=0, color="red", linestyle="--", linewidth=2)
plt.xlabel("X Values")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted Values")
plt.grid()

plt.subplot(2, 1, 2)
plt.hist(residuals, bins=30, edgecolor="black", alpha=0.7)
plt.xlabel("Residuals")
plt.show()
````

Notice how the histogram shows errors that are not normal. This means that the p-values and confidence intervals for the coefficients are only an approximation. This approximation is good in this case due to the large sample size. The residuals plot also indicates non-linearity, in particular, it shows quadratic dependence. 

Now let's run the model again using the $X^2$ term and observe how this impacts the performance of the model and the $R^2$ value.


````python
# Fit a quadratic model
y = Y
X_matrix = np.column_stack([X, X**2])
X_cons = sm.add_constant(X_matrix)

# Fit the model
est = sm.OLS(y, X_cons).fit()

# Print results
print(est.summary())

# Plot quadratic fit
X_fit = np.linspace(X.min(), X.max(), 200)
X_fit_matrix = np.column_stack([X_fit, X_fit**2])
X_fit_cons = sm.add_constant(X_fit_matrix)
Y_fit_pred = est.predict(X_fit_cons)

plt.figure(figsize=(10, 6))
plt.scatter(X, values, label="Data Points", color="blue")
plt.plot(X_fit, Y_fit_pred, color="red", linewidth=3, label="Quadratic Fit")
plt.xlabel("X Value")
plt.ylabel("Y Value")
plt.title("Regression: X vs Y")
plt.legend()
plt.grid()
plt.show()
````

We also plot the residuals using this model, observe how the model is improving.

````python
residuals = est.resid

plt.figure(figsize=(10, 6))
plt.subplot(2, 1, 1)
plt.scatter(X, residuals, alpha=0.6, color="blue")
plt.axhline(y=0, color="red", linestyle="--", linewidth=2)
plt.xlabel("X Values")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted Values")
plt.grid()

plt.subplot(2, 1, 2)
plt.hist(residuals, bins=30, edgecolor="black", alpha=0.7)
plt.xlabel("Residuals")
plt.show()
````

Now we will do another simulation to consider hetero-skedasticity. We will simulate $Y=a+bX+U$ where $U$ is normally distributed with mean $0$ and standard deviation $X$. First we will run the linear regression model and plot a scatterplot of our data.

````python
# Simulate hetero-skedasticity
sample_size = 500
rng = np.random.default_rng()
X = rng.uniform(0, 5, sample_size)
U = []
[U.append(rng.normal(0, x)) for x in X]
a = 10
b = 2
Y = a + b * X + U

y = Y
X_cons = sm.add_constant(X)

# Fit the model
est = sm.OLS(y, X_cons).fit()

# Print results
print(est.summary())

values = Y
estimated_values = est.predict(X_cons)

plt.figure(figsize=(10, 6))
plt.scatter(X, values, label="Data Points", color="blue")
plt.plot(X, estimated_values, color="red", label="Regression Line")
plt.xlabel("X Value")
plt.ylabel("Y Value")
plt.title("Linear Regression: X vs Y")
plt.legend()
plt.grid()
plt.show()
````

The next step is to perform residual analysis which should reveal the hetero-skedasticity.

````python
residuals = est.resid

plt.figure(figsize=(10, 6))
plt.subplot(2, 1, 1)
plt.scatter(X, residuals, alpha=0.6, color="blue")
plt.axhline(y=0, color="red", linestyle="--", linewidth=2)
plt.xlabel("X Values")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted Values")
plt.grid()

plt.subplot(2, 1, 2)
plt.hist(residuals, bins=30, edgecolor="black", alpha=0.7)
plt.xlabel("Residuals")
plt.show()
````

The hetero-skedasticity can be reduced by applying a transformation to $Y$ such as $ln(Y)$ or $\sqrt{Y}$. Try both transformations on $Y$ and run the code below to observe how the model and r-squared value changes.

````python
# Apply transformation to Y
Y = np.log(Y)

# Fit the model again
est = sm.OLS(Y, X_cons).fit()

# Print results
print(est.summary())

values = Y
estimated_values = est.predict(X_cons)

# Plot the transformed data with regression line
plt.figure(figsize=(10, 6))
plt.scatter(X, values, label="Data Points", color="blue")
plt.plot(X, estimated_values, color="red", label="Regression Line")
plt.xlabel("X Value")
plt.ylabel("Y Value")
plt.title("Linear Regression: X vs Y")
plt.legend()
plt.grid()
plt.show()

# Perform residual analysis
residuals = est.resid

plt.figure(figsize=(10, 6))
plt.subplot(2, 1, 1)
plt.scatter(X, residuals, alpha=0.6, color="blue")
plt.axhline(y=0, color="red", linestyle="--", linewidth=2)
plt.xlabel("X Values")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted Values")
plt.grid()

plt.subplot(2, 1, 2)
plt.hist(residuals, bins=30, edgecolor="black", alpha=0.7)
plt.xlabel("Residuals")
plt.show()
````
